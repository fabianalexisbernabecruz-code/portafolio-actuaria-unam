## **Búsqueda Secuencial**

</font>

<font face="Times New Roman">

El algoritmo de búsqueda secuencial, también conocido como búsqueda lineal, es una técnica sencilla y fácil de implementar para encontrar un elemento específico en una estructura de datos, como una lista o un vector. Su enfoque consiste en examinar cada elemento de la colección de forma secuencial hasta encontrar el dato que estamos buscando.

Lo notable de este algoritmo es que no requiere ningún requisito especial y funciona eficazmente sin importar la longitud de la colección o el orden en el que se encuentren los elementos. Esto significa que puede utilizarse en una amplia variedad de situaciones, lo que lo convierte en una herramienta versátil para la búsqueda de datos.


Inicio: El algoritmo toma como entrada una colección de datos, y selecciona un elemento o dato a buscar.

Comparación: Luego, compara el elemento seleccionado con los demás datos.

Coincidencia: Si el elemento actual es igual al valor buscado, el algoritmo ha encontrado el elemento y puede finalizar. (Devuelve un valor true)

No coincidencia: Si el elemento actual no coincide con el valor buscado, el algoritmo avanza al siguiente elemento de la colección. Esto se logra moviendo el marcador o puntero al siguiente elemento (posición 1, luego 2, y así sucesivamente).

Repetición: Los pasos 2-4 se repiten hasta que se encuentre el valor buscado o se haya recorrido toda la colección sin éxito.

Resultado: Si el algoritmo llega al final de la colección sin encontrar el valor, informa que el elemento no se encuentra en la colección o retorna un indicador específico de no coincidencia.

## **Búsqueda Binaria**

</font>

<font face="Times New Roman">

Este algoritmo da una mejor manera para encontrar un elemento buscado en una lista, necesariamente ordenada, pues en tan solo una "pasada" ya descarta una gran cantidad de elementos que no pueden ser el elemento buscado, a diferencia del algoritmo anterior que, en el peor de los casos, debe recorrer toda la lista.

Basicamente lo que hace el algoritmo es tomar el dato central de la lista y compararlo con el dato a buscar, de serlo devuevlve la posición de este y si no es se tiene que el dato central es mayor o en otro caso menor y descarta la mitad que ya no tenga sentdio considerar pues lal ista está ordenada y vuelve a hacer lo mismo hasta que el valor central de la lista sea el dato o hasta que el dato central no sea el dato buscado y ya no haya más elementos a comparar.

Veamos el paso a paso del algoritmo para una lista de $n$ elementos.

Lo primero que debemos saber es la longitud de la lista y el dato a buscar en la misma para ahora pasar al algoritmo.

1. Si la lista tiene $0$ elementos devolvemos un $-1$, ( aquí vamos a considerar esto, para indicar que no está el elemento en la lista proporcionada, pero se puede usar otra cosa como "el elemento no está en la lista"), desde que no hay elementos en la lista.

2. Si tiene un solo elemento lo comparamos con el que deseamos buscar y de serlo devolvemos el índice $0$ (la lista solo tiene un elemento; la posición de éste debe ser $0$ (python)) de lo contrario devolvemos $-1$ según corresponda.


3. Si no sucede los anteriores casos, $i.e.$ la lista tiene al menos 2 elementos procedemos como sigue:

4. Consideramos como segmento inicial a la lista completa, tomamos el dato central de la lista y lo comparamos con el dato buscado, si es devolvemos el indice del dato central.

5. Si el dato central no es el dato buscado, entonces descartamos todo el segmento que no tenga sentido considerar más, por ejemplo si el dato central fuera mayor que el dato buscado entonces todos los datos después del central y él mismo no pueden ser el dato buscado dado que la lista está ordenada y si fuera menor el dato central al buscado, no tiene sentido buscar en los elementos menores que el dato central.

6. Consideramos ahora el segmento que va de un dato después o antes (según sea el caso) del dato central hasta el dato final o inicial (según el caso) y volvemos a obtener el dato central de este segmento nuevo; resultante de descartar el segmento donde no tiene sentido buscar.

7. Repetimos de 1 a 6 pasos hasta que el dato buscado sea idéntico al central o no lo sea y en este caso decimos que no estaba ese dato buscado en la lista.

Este peor caso se toca cuando la última iteración se hace sobre una lista de longitud 1, un solo hay un elemento en ella, esto es cuando el elemento nunca coincide con el medio tomado entre cada lista salvo cuando se llega a una lista de longitud 1 en la que puede o no ser ese el dato que buscamos. Lo anterior se da pues con cada iteración del bucle while, el algoritmo reduce a la mitad el número de elementos a buscar ya que actualiza izquierda o derecha para ser centro + 1 o centro - 1, respectivamente, eliminando con ello la mitad de la lista de la búsqueda en la que no tiene sentido buscar más. Por lo tanto, el número de elementos a buscar se reduce a la mitad en cada paso, lo que significa que el número de pasos necesarios para encontrar un elemento (o determinar que no está en la lista) es el logaritmo base 2 de $n$. Por lo tanto, la complejidad de tiempo es $O(log_{2}n)$.

### **Recursiva**

</font>

<font face="Times New Roman">

#### **Respecto al tiempo**

</font>

<font face="Times New Roman">

Vamos por casos para construir una función de recursión. Para el caso cuando se requiere buscar un elemento en una colección (lista) de tamaño 0, solo se realiza una comparación que toma tiempo constante $b$ (devolver -1) y si la lista es de tamaño 1, toma 2 operaciones pues comparamos con un solo elemento y devolvemos su indice si es el buscado o -1 si no es así.

Para el caso de una colección de tamaño 2 ó mayor y el centro no es igual al dato buscado tenemos en cualquier caso una llamada recursiva, y es necesario partir a la mitad dicha colección y preguntar si el elemento se encuentra en la mitad de la colección, en caso de no ser así es, se verifica si el elemento que se busca es mayor o menor al elemento que se encuentra a la mitad de la colección. Y dependiendo de la comparación se descarta una de las 2 mitades y se procede a buscar al elemento en la otra mitad. Los pasos se repiten hasta que para alguna mitad o centro coincida con el elemento o hasta que solo quede una lista de un elemento sin descartar y por tanto se haga una comparación en la que puede o no ser el elemento del centro o mitad (el único que queda) igual al dato buscado. De lo anterior se deriva que buscar un elemento en una colección de tamaño $n$ tomará en total $T(\frac{n}{2})+a$ para $a$ la contante que representa el tiempo que le toma al dispositivo de cómputo realizar la comparación para saber si el elemento buscado se encuentra es o no la mitad o centro tomado.
Tenemos entonces que la función de recurrencia es:

$$T(n)=\begin{cases}
b, & n=0\,\acute{o}\,n=1\\
T(\frac{n}{2})+a & n\geq2
\end{cases}$$

Luego entonces sin pérdida de generalidad y para que el calculo sea lo más simple posible supondremos que la lista es de tamaño $2^k$, es decir, par, si fuera de tamaño impar un numero impar digamos $s$, entonces $ \exists l \in \mathbb{N}$ tal que $s < 2^l$, $i.e.$ basta acotar los pares como haremos.

$Demostracion$

Sea $n=2^k$ la cantidad de elementos que tiene la lista en la cual se va a buscar el elemento.
- Al final de la primer iteración la longitud a tratar es de $ \frac{2^k}{2}=2^{k-1}$
- Para la segunda es $\frac{2^{k-1}}{2}=2^{k-2}$
- Para la tercera $\frac{2^{k-2}}{2}=2^{k-3}$
- ...
- Por el patron de comportamiento para la $i-esima$ iteración tenemos que es $\frac{2^{k-i}}{2}=2^{k-i}$
- Si $i=k$ para llegar a longitud de lista unitaria llegamos a lo que sigue $\frac{2^{k-k}}{2}=2^{k-k}$ esto es $2^0=1$

Por lo anterior ya vimos que el programa hace aproximadamente $k$ comparaciones para llegar a segmentos unitarios, luego entonces obteniendo $k$ del siguiente despeje:

Si $$n=2^k \implies log_{2}{n}=log_{2}{2^k}\implies log_{2}{n}=k \cdot log_{2}{2}=k \cdot 1=k $$

$$\therefore  log_{2}{n}=k$$

Se sigue que el algoritmo hace aproximadamente $log_{2}{n}$, aún si multiplicamos por una constante esta función se sigue manteniendo que el orden de complejidad del algoritmo es $log_{2}{n}$ así que si $T(n)$ es la función de tiempo

$$\therefore T(n)\in O(log_{2}{n})$$

Sea $T(n)$ la función de expacio que le toma al algoritmo almacenar lo que necesita para encontrar un elemento en una lista ordenada.

Como para cualquier tamaño de lista hace exactamente 3 asignaciones de memoria para las variables "izquierda", "derecha" y "centro" y con ellas trabaja el resto del algortimo modificandolas, cambiandolas, pero *in situ*, se sigue que suponiendo que el espacio que le requiere es alguna constante $k$ para cada asignación entonces $\forall n \in \mathbb{N}$ tenemos que $T(n)=3k$ pues son 3 las asignaciones de espacio $k$ (para alguna $k$ constante), lo que dice es que, tomando,  por notación de Big O o definición de cota superior asintótica, $c=3$ y $g(n)=k$, entonces $T(n)\in O(g(n))$

$$\therefore T(n)\in O(1)$$

### **Recursiva**

</font>

<font face="Times New Roman">

#### **Respecto al tiempo**

<font face="Times New Roman">

Al igual que en la anterior versión la función de recurrencia para el algoritmo es

$$T(n)=\begin{cases}
b, & n=0\,\acute{o}\,n=1\\
T(\frac{n}{2})+a & n\geq2
\end{cases}$$


Sea $T(n)$ el número de operaciones que le toma al algoritmo buscar un dato en una lista de tamaño $n$ y dada la función de recurrencia.

$P.D.\ T(n)=b+a*\log_2 n$.

Cuanfo inicia el algoritmo para por comparar con el centro esta operación es constante digamos $a$ y luego pasa, por ser el peor caso a una llamada recursiva de la función, dicha función es $T(n)$ aspi que basta mostrar que dada la función de recurrencia se cumple que $T(n)=b+a*\log_2 n$.

Sea $n=2^k$ la cantidad de elementos que tiene la lista en la cual se va a buscar el elemento.

$$\begin{eqnarray*}
T(n)    & = &   T(2^{k})\\
  & = & (T(2^{k-1})+a)\\
  & = & (T(2^{k-2})+a)+a\\
  & = & (T(2^{k-3})+a)+a+a\\
i-veces & \vdots & \\
  & = & T(2^{k-i})+i \cdot a\\
i=k     & \vdots & \\
  & = & T(2^{k-k})+k \cdot a\\
  & = & T(2^{0})+ k\cdot a\\
  & = & T(1)+ k\cdot a\\
  & = & b+ a\cdot log_{2}(n)\\
  \end{eqnarray*}$$

  
Las igualdades vienen dadas por lo siguiente:
* la primera es por conveniencia de ver las divisiones de la lista, esto es cuando definimos $n=2^k$, lo sustituimos.
* la siguiente igualdad es por definición de $T(n)$,
* la tercer igualdad viene dada por la definición de $T(n)$ aplicada a $2^{k-1}$
* la cuarta dada por la definición de $T(n)$ aplicada a $2^{k-2}$
* Lo de $i-veces$ es por el claro el patrón que se forma en cuanto a las potencias de 2 y la cantidad de constantes $a$ que hay directamente proporcionales a la cantidad de divisiones de $2^{k}$
* lo de $i=k$ sucede porque haciendo esto llegamos al caso base, aquel que ya no usa la recursividad cosa que es típica de los algoritmos recursivos, se debe llegar a algo que podamos saber qué hacer y en este caso cuánto tiempo tomó.
* La que sigue es reemplazo de $i=k$,
* el resto es simple algebra elemental.
* finalmente reemplazo de $k$ que se sigue de:
Si $$n=2^k \implies log_{2}{n}=log_{2}{2^k}\implies log_{2}{n}=k \cdot log_{2}{2}=k \cdot 1=k $$

$$\therefore  log_{2}{n}=k$$

Ya que $a$ y $b$ son constantes podemos concluir que, la función que 'crece' más rápido es $a\cdot\log_{2}n$

Finalmente si tomamos $a=c$ y $g(n)=log_{2}n$, entonces se puede concluir que

$$\therefore T(n)\in O(\log_{2}n)$$

</font>

<font face="Times New Roman">

Para esta el orden de complejidad del almacenamiento es $O(\log_{2}n)$ cuya demostración no es muy formal pero tiene como principio el siguiente razonamiento sustentado en el análisis previo.

Como usa algunas variables como el centro que lo define en términos de los índices de los elementos primero y último tenemos que usa sola llamada del algoritmo sin recursión tiene orden de complejidad $O(1)$ como vimos para la versión iterativa, el problema, o en lo que difiere la versión iterativa de la recursiva es que esta versión hace llamadas a la función en sí misma, entonces tenemos que consume tanta memoria como pilas de llamadas de sí misma hace, esto es, como vimos por la demostración de la versión iterativa que al algoritmo en su versión iterativa le toma $k=log_{2}(n)$ comparaciones (tambiém en esta versión), en este caso almacenamiento de variables encontrar un elemento en la lista con una lista de tañaño $n$, pues cada que se llama la función se hace un alamcenamiento de las variables iniciales, se sigue que $T(n)=O(1) \cdot k = O(1) \cdot log_{2}(n)$ y así, por teoremas de orden de complejidad multiplicar por una constante a la función no altera su orden de complejidad, llegamos de todo lo anterior a concluir que:

$$\therefore T(n)\in O(\log_{2}n)$$

## **Búsqueda Secuencial**

</font>

<font face="Times New Roman">

### **Iterativa**

</font>

```python
def BusquedaSecuencialIterativa(lista, elemento):
    """
    Realiza una búsqueda secuencial iterativa en una lista para encontrar un elemento.
    Esta función busca un elemento en una lista recorriendo cada elemento de la lista
    en un bucle for y comparando con el valor buscado hasta encontrar una coincidencia o
    recorrer toda la lista.
    Parameters
    ----------
    lista : list
        La lista en la que se va a buscar el elemento.
    elemento : cualquier tipo de dato
        El elemento que se desea encontrar en la lista.
    Returns
    -------
    int
        El índice en el que se encuentra el elemento si se encuentra; de lo contrario, -1.
    """
    for i in range(len(lista)):
        if lista[i] == elemento:
            return i  # Devuelve el índice si encuentra el elemento
    return -1  # Devuelve -1 si el elemento no está en la lista

# Fin del programa
```

<font face="Times New Roman">

### **Recursiva**

</font>

```python
def BusquedaSecuencialRecursiva(lista, elemento, indice=0):
    """
    Realiza una búsqueda secuencial recursiva en una lista para encontrar un elemento.
    Esta función busca un elemento en una lista de manera recursiva comparando cada
    elemento de la lista con el valor buscado hasta encontrar una coincidencia o
    recorrer toda la lista.

    Parameters
    ----------
    lista : list
        La lista en la que se va a buscar el elemento.
    elemento : cualquier tipo de dato
        El elemento que se desea encontrar en la lista.
    indice : int, optional
        El índice actual en el que se está buscando (por defecto, 0).

    Returns
    -------
    int
        El índice en el que se encuentra el elemento si se encuentra; de lo contrario, -1.
    """
    if indice == len(lista):
        return -1  # El elemento no se encontró en la lista

    if lista[indice] == elemento:
        return indice  # Devuelve el índice si encuentra el elemento

    return BusquedaSecuencialRecursiva(lista, elemento, indice + 1)

# Fin del programa
```

## **Búsqueda Binaria**

</font>

<font face="Times New Roman">

### **Iterativa**

</font>

```python
def BusquedaBinariaIterativa(lista, dato):
    """
    Busca un elemento llamado dato en la lista y devuelve el índice donde está (respecto a la lista) o -1 si no está en la lista.
    Parameters
    ----------
    lista : list
        La lista en la que se va a buscar el elemento ordenada en forma no decreciente.
    dato : cualquiera que permita darlo en una lista ordenada como int, float, str...
        El elemento que se desea buscar en la lista.
    Returns
    -------
    centro : int
        El índice del elemento en la lista si se encuentra; de lo contrario, retorna -1.
    """
    izquierda = 0                           # Tomamos como segmento a la lista completa y el primer elemento en índice es 0
    derecha = len(lista) - 1                # El último elemento es en índice el número total de la lista menos uno ya que inicia en 0 los índices
    centro = (izquierda + derecha) // 2     # Calculamos el centro, usamos divisón entra pues no hay índices fraccionario
    while izquierda <= derecha:             # Mientra la izquierda no sea mayor que la derecha
        if lista[centro] == dato:           # Comparamos el centro con el dato
            return centro                   # De ser iguales lo devolvemos
        elif lista[centro] < dato:          # De no cumplirse lo anterior
            izquierda = centro + 1          # Actualizamos el valor de la izquiera, eliminando así el segmento donde no tiene sentido buscar al dato
        else:                               # Si no pasan los otros, debe de pasar este
            derecha = centro - 1            # Actualizamos el valor de la derecha...
        centro = (izquierda + derecha) // 2 # Actualizamos el centro para repetirlo todo
    return -1                               # Devolvemos -1; no estaba el dato en la lista y se produce cuando izquierda > derecha

# Fin del programa
```

<font face="Times New Roman">

### **Recursiva**

</font>

```python
def BusquedaBinariaRecursiva(lista, dato, izquierda, derecha):
    """
    Busca un elemento llamado dato en la lista y devuelve el índice donde está (respecto a la lista) o -1 si no está en la lista.
    Parameters
    ----------
    lista : list
        La lista en la que se va a buscar el elemento ordenada en forma no decreciente.
    dato : cualquier tipo comparable con los elementos de la lista (int, float, str, etc.).
        El elemento que se desea buscar en la lista.
    izquierda : int
        El índice de inicio de la porción actual de la lista en la que buscar (generalmente 0).
    derecha : int
        El índice del último elemento de la lista (len(lista) - 1).
    Returns
    -------
    centro : int
        El índice del elemento en la lista si se encuentra; de lo contrario, retorna -1.
    """
    if izquierda > derecha:                # Caso base: si el índice izquierda es mayor que el índice derecha, el elemento no está en la lista.
        return -1                          # Devolvemos -1 indicando que el elemento no estpa
    centro = (izquierda + derecha) // 2    # Calcula el índice medio de la porción actual de la lista.
    if lista[centro] == dato:              # Si el elemento en el índice medio es igual al dato, lo hemos encontrado.
        return centro                      # Devolvemos el índice del centro que es el mismo que el del dato
    elif lista[centro] < dato:             # Si el dato es menor que el elemento en el índice medio, busca en la mitad derecha de la lista de forma recursiva.
        return BusquedaBinariaRecursiva(lista, dato, centro + 1, derecha)   # Pasamos los parámetros nuevos de manera implícita
    else:                                  # Si no sucede cualquiera de los anteiores debe pasar este y busca en la mitad izquierda de la lista de forma recursiva.
        return BusquedaBinariaRecursiva(lista, dato, izquierda, centro - 1) # Pasamos los parámetrso nuevos de manera implícita

# Fin del programa
```

```python
def selectionSort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    n = len(lista)

    for i in range(n):
        # Suponemos que el primer elemento sin ordenar es el mínimo
        minimo = i

        # Buscamos el elemento más pequeño en la parte no ordenada
        for j in range(i + 1, n):
            if lista[j] < lista[minimo]:
                minimo = j

        # Intercambiamos el elemento más pequeño con el primer elemento sin ordenar
        lista[i], lista[minimo] = lista[minimo], lista[i]

    return lista

#### FIN DEL PROGRAMA
```

```python
def insertionSort(arr):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    for i in range(1, len(arr)):# Comenzamos desde el segundo elemento (índice 1)
        key = arr[i]# Guardamos el elemento actual que queremos insertar en su posición correcta
        j = i - 1# Inicializamos un índice para comparar con los elementos anteriores
                    # Comparamos el elemento actual con los elementos anteriores y desplazamos los elementos mayores hacia la derecha
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
                # Colocamos el elemento actual en su posición correcta
        arr[j + 1] = key

    return arr

#### FIN DEL PROGRAMA
```

```python
# implementacion in-situ del algoritmo burbuja
def bubbleSort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    #asigno el largo de la lista a una variable
    n = len(lista)
    #bandera para identificar cambios
    intercambio = True
    #variable para saber cuantos han sido ordenados
    j = 0
    #si hay cambios entro al ciclo
    while(intercambio):
        #los j-esimos elementos ya han sido ordenados
        j = j+1
        #asumimos que no hay intercambios
        intercambio = False
        #con este for se comparan las parejas que no estan ordenadas
        for i in range(n-j):
            if lista[i] > lista[i+1] :
                # intercambio en una linea (swap)
                lista[i], lista[i+1] = lista[i+1], lista[i]
                intercambio = True
    #devuelve la lista solo para ser formales IN-SITU
    return lista

#### FIN DEL PROGRAMA
```

```python
def merge_sort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    ##creamos el caso trivial, cuando no hay más qué hacer
    if len(lista) < 2:
        return lista
    ##implementamos el caso cuanto la lista tiene más de dos elemento
    else:
        medio = len(lista) // 2
        izq = merge_sort(lista[:medio])
        der = merge_sort(lista[medio:])
        return merge(izq, der)
#### FIN DEL PROGRAMA

### creamos la funcion que nos permite hacer el paso de mezclar las
## listas una vez que han sido divididas en el paso anterior
def merge(xs, ys):
    """Fusiona dos listas ordenadas en una sola lista ordenada, es una función auxiliar.

    Parameters
    ----------
    xs (list): La primera lista ordenada.
    ys (list): La segunda lista ordenada.
    xs : list
    ys : list

    Returns
    -------
     Una nueva lista que contiene todos los elementos de xs e ys, ordenados de manera ascendente : list.

    """
    ix, iy = 0, 0 # Inicializamos los índices para recorrer xs e ys.
    result = []  #Inicializamos una lista vacía para almacenar el resultado.
    while ix < len(xs) and iy < len(ys):
        if xs[ix] < ys[iy]:
            result.append(xs[ix])
            ix += 1
        else:
            result.append(ys[iy])
            iy += 1
    # Después de salir del bucle, agrega los elementos restantes (si los hay) de xs e ys.
    result += xs[ix:]
    result += ys[iy:]
    return result

#### FIN DEL PROGRAMA
```

```python
def quicksort(arr):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    # Paso base: Si la lista está vacía o tiene un solo elemento, ya está ordenada.
    if len(arr) <= 1:
        return arr

    # Elegimos un elemento pivote (generalmente el primer elemento en este caso).
    pivote = arr[0]

    # Creamos tres listas vacías para los elementos menores, iguales y mayores que el pivote.
    menor, igual, mayor = [], [], []

    # Recorremos la lista y distribuimos los elementos en las tres listas según su relación con el pivote.
    for elemento in arr:
        if elemento < pivote:
            menor.append(elemento)
        elif elemento == pivote:
            igual.append(elemento)
        else:
            mayor.append(elemento)

    # Llamamos recursivamente a quicksort en las listas menor y mayor.
    # Concatenamos las tres listas en el orden correcto para obtener la lista ordenada.
    return quicksort(menor) + igual + quicksort(mayor)

#### FIN DEL PROGRAMA
```

```python
# creamos una función para medir el tiempo y guardar los tiempos en una lista que después pasaremos como parametro para crear las gráficas
def MedidorTiempo(algoritmo, listaX, resultados):
    """
    Mide el tiempo que le toma al algoritmo ordenar una lista
    Toma el algoritmo y la lista a ordenar y la lista donde guardará la lista resultados
    Regresa una lista con los tiempos que le tomó al algoritmo ordenar la lista
    """
    listaXcopia = listaX.copy()
    inicio = time.time()
    algoritmo(listaXcopia)
    fin = time.time()
    tiempo = fin - inicio
    resultados.append(tiempo)

    return resultados
```
