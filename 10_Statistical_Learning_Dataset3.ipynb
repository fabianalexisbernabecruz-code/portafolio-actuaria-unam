**Fuente:** Sistema Nacional de Informacion e Indicadores de Vivienda — SEDATU/CNBV

**Motivacion actuarial:** Modelar el monto de credito hipotecario es la base para el calculo de reservas tecnicas en seguros de vida con componente hipotecario (deudores), estimacion de perdida esperada y modelos de pricing inmobiliario.

**Reto tecnico:** distribucion muy sesgada a la derecha (lognormal/Gamma), heterogeneidad regional brutal (CDMX vs Oaxaca), requiere transformacion de la respuesta y comparacion de funcion de enlace en MLG.

**Pipeline:**
1. MLG con familia Gamma/Tweedie (funcion de enlace log).
2. Arboles de decision (regresion).
3. SVR (Support Vector Regression).
4. K-Vecinos mas Cercanos (KNN regresion).
5. Redes Neuronales — MLP shallow (regresion).

La funcion `cargar_sniiv` se encarga de descargar el conjunto de datos desde la API del SNIIV. Intenta la API directa para multiples anios, luego genera un dataset sintetico basado en estadisticas publicas del SNIIV como fallback.
</font>

```python
import pandas as pd
import numpy as np
import os

DATA_API_BASE = "https://sniiv.sedatu.gob.mx/api/CuboAPI/GetFinanciamiento/"
LOCAL_CSV = "sniiv_financiamiento.csv"
RANDOM_STATE = 42

def cargar_sniiv(sample_n=100_000):
    df = None
    frames = []
    for year in range(2020, 2025):
        try:
            url = f"{DATA_API_BASE}{year}"
            req = Request(url, headers={'User-Agent': 'Mozilla/5.0'})
            with urlopen(req, timeout=30) as resp:
                raw = resp.read().decode('utf-8')
            data = json.loads(raw)
            if isinstance(data, list) and len(data) > 0:
                frames.append(pd.DataFrame(data))
                print(f"  Anio {year}: {len(frames[-1])} registros")
            elif isinstance(data, dict) and 'data' in data:
                frames.append(pd.DataFrame(data['data']))
                print(f"  Anio {year}: {len(frames[-1])} registros")
            else:
                print(f"  Anio {year}: respuesta vacia o formato inesperado")
        except Exception as e:
            print(f"  Anio {year} fallo: {e}")
    if frames:
        df = pd.concat(frames, ignore_index=True)
        print(f"  API SNIIV: {len(df)} registros totales")

    if df is None or len(df) == 0:
        print("API SNIIV no disponible. Generando dataset sintetico basado en estadisticas publicas SNIIV...")
        np.random.seed(RANDOM_STATE)
        n = 150_000
        estados = ['Aguascalientes','Baja California','Baja California Sur','Campeche',
                    'Coahuila','Colima','Chiapas','Chihuahua','Ciudad de Mexico',
                    'Durango','Guanajuato','Guerrero','Hidalgo','Jalisco',
                    'Mexico','Michoacan','Morelos','Nayarit','Nuevo Leon',
                    'Oaxaca','Puebla','Queretaro','Quintana Roo','San Luis Potosi',
                    'Sinaloa','Sonora','Tabasco','Tamaulipas','Tlaxcala',
                    'Veracruz','Yucatan','Zacatecas']
        organismos = ['INFONAVIT', 'FOVISSSTE', 'BANCA', 'OTROS']
        tipos_viv = ['Nueva', 'Usada', 'Mejoramientos', 'Otros']
        rangos_edad = ['18-25', '26-30', '31-35', '36-40', '41-45',
                       '46-50', '51-55', '56-60', '61-65', '66+']
        prob_edo = np.array([0.02,0.05,0.015,0.01,0.04,0.01,0.025,0.04,
                             0.12,0.015,0.06,0.02,0.025,0.08,0.09,0.03,
                             0.02,0.015,0.07,0.02,0.04,0.03,0.025,0.02,
                             0.03,0.03,0.015,0.03,0.01,0.04,0.025,0.015])
        prob_edo = prob_edo / prob_edo.sum()
        prob_org = np.array([0.55, 0.15, 0.25, 0.05])
        prob_viv = np.array([0.45, 0.30, 0.20, 0.05])

        estado = np.random.choice(estados, n, p=prob_edo)
        organismo = np.random.choice(organismos, n, p=prob_org)
        tipo_vivienda = np.random.choice(tipos_viv, n, p=prob_viv)
        rango_edad = np.random.choice(rangos_edad, n,
                                       p=[0.08,0.15,0.20,0.18,0.14,0.10,0.07,0.04,0.025,0.015])
        anio = np.random.choice([2020,2021,2022,2023,2024], n, p=[0.18,0.19,0.21,0.21,0.21])
        acciones = np.random.poisson(lam=3, size=n) + 1

        base_monto = {'INFONAVIT':350000,'FOVISSSTE':450000,'BANCA':1200000,'OTROS':600000}
        factor_edo = {
            'Ciudad de Mexico':2.5,'Nuevo Leon':1.8,'Jalisco':1.5,'Queretaro':1.6,
            'Baja California Sur':1.7,'Quintana Roo':1.4,'Baja California':1.3,
            'Chihuahua':1.1,'Sonora':1.1,'Coahuila':1.1,'Tamaulipas':1.0,
            'Guanajuato':1.0,'Aguascalientes':1.1,'Colima':1.0,'Sinaloa':1.0,
            'Mexico':1.2,'Morelos':1.1,'Puebla':0.9,'San Luis Potosi':0.9,
            'Yucatan':1.0,'Nayarit':0.9,'Hidalgo':0.85,'Durango':0.85,
            'Tabasco':0.85,'Veracruz':0.8,'Michoacan':0.8,'Zacatecas':0.8,
            'Campeche':0.85,'Tlaxcala':0.75,'Guerrero':0.7,'Oaxaca':0.65,'Chiapas':0.6}
        factor_viv = {'Nueva':1.3,'Usada':1.0,'Mejoramientos':0.4,'Otros':0.7}
        factor_edad_map = {'18-25':0.7,'26-30':0.9,'31-35':1.0,'36-40':1.05,
                           '41-45':1.1,'46-50':1.05,'51-55':1.0,'56-60':0.95,
                           '61-65':0.9,'66+':0.85}

        monto = np.array([base_monto[org]*factor_edo.get(edo,1.0)*factor_viv[viv]*factor_edad_map[edad]
                          for org,edo,viv,edad in zip(organismo,estado,tipo_vivienda,rango_edad)])
        noise = np.random.lognormal(0, 0.4, n)
        monto = (monto * noise).astype(int)
        monto = np.clip(monto, 50_000, 15_000_000)

        estado_list = list(estado)
        tipo_list = list(tipo_vivienda)
        for i in np.random.choice(n, int(n*0.02), replace=False):
            estado_list[i] = np.nan
        for i in np.random.choice(n, int(n*0.01), replace=False):
            tipo_list[i] = np.nan

        df = pd.DataFrame({'anio':anio,'estado':estado_list,'organismo':organismo,
                           'tipo_vivienda':tipo_list,'rango_edad':rango_edad,
                           'monto':monto,'acciones':acciones})
        print(f"  Dataset sintetico generado: {df.shape}")

    if df is None and os.path.exists(LOCAL_CSV):
        print(f"Usando archivo local: {LOCAL_CSV}")
        df = pd.read_csv(LOCAL_CSV, low_memory=False)

    if df is None:
        raise RuntimeError("Todas las rutas de carga fallaron.")

    print(f"  shape crudo: {df.shape}")
    df.columns = [str(c).strip().lower().replace(" ", "_") for c in df.columns]
    if sample_n and len(df) > sample_n:
        df = df.sample(sample_n, random_state=RANDOM_STATE).reset_index(drop=True)
        print(f"  Subsample: {df.shape}")
    return df

df = cargar_sniiv(sample_n=100_000)
print(f"\nDataset cargado: {df.shape[0]:,} filas x {df.shape[1]} columnas")
```

La funcion `reporte_faltantes` cuenta los nulos por columna. El patron de faltantes se clasifica como MAR (Missing At Random) porque las columnas geograficas/descriptivas son las afectadas y su ausencia depende de variables observadas.

```python
def reporte_faltantes(df):
    na = df.isna().sum()
    pct = (na / len(df) * 100).round(3)
    out = pd.DataFrame({'n_missing': na, 'pct_missing': pct})
    out = out[out.n_missing > 0].sort_values('pct_missing', ascending=False)
    if len(out) == 0:
        print('Sin valores faltantes.')
    else:
        display(out)
    return out

faltantes = reporte_faltantes(df)
```

<font face="Times New Roman">

Se procesan las variables del dataset: los rangos de edad se convierten a punto medio numerico para poder usarlos como variable continua, y se identifican las columnas continuas y categoricas.
</font>

```python
def parse_rango_edad(s):
    if pd.isna(s): return np.nan
    s = str(s).strip()
    if '+' in s:
        num = ''.join(c for c in s if c.isdigit())
        return float(num) + 2.5 if num else np.nan
    parts = [p.strip() for p in s.replace('-', ' ').split() if p.strip().isdigit()]
    if len(parts) == 2: return (float(parts[0]) + float(parts[1])) / 2
    elif len(parts) == 1: return float(parts[0])
    return np.nan

col_edad = next((c for c in df.columns if 'edad' in c.lower()), None)
if col_edad:
    df['edad_midpoint'] = df[col_edad].apply(parse_rango_edad)
    print(f'Rango de edad parseado: {col_edad} -> edad_midpoint')
    print(f'  Rango: {df["edad_midpoint"].min():.0f} - {df["edad_midpoint"].max():.0f}')

cols_num = df.select_dtypes(include=[np.number]).columns.tolist()
cols_num = [c for c in cols_num if c != col_monto and c != 'monto']
cols_cat = df.select_dtypes(include=['object']).columns.tolist()
print(f'\nVariables continuas ({len(cols_num)}): {cols_num}')
print(f'Variables categoricas ({len(cols_cat)}): {cols_cat}')
```

## Parte 2 — Feature Engineering y Transformaciones ``

Las transformaciones de variables buscan (i) estabilizar la varianza de la respuesta para modelos que asumen homocedasticidad, (ii) reducir la influencia de outliers, y (iii) crear features derivadas con interpretacion actuarial.
</font>

```python
def imputar(df, cols_num, cols_cat, metodo='mediana'):
    df = df.copy()
    if metodo == 'mediana':
        df[cols_num] = df[cols_num].fillna(df[cols_num].median())
    elif metodo == 'knn':
        from sklearn.impute import KNNImputer
        imp = KNNImputer(n_neighbors=5)
        df[cols_num] = imp.fit_transform(df[cols_num])
    for c in cols_cat:
        if c in df.columns:
            df[c] = df[c].fillna(df[c].mode().iloc[0] if len(df[c].mode()) > 0 else 'DESCONOCIDO')
    return df

cols_num_imp = [c for c in cols_num if df[c].isna().any()]
cols_cat_imp = [c for c in cols_cat if df[c].isna().any()]
print(f'Cols numericas a imputar: {cols_num_imp}')
print(f'Cols categoricas a imputar: {cols_cat_imp}')
df = imputar(df, cols_num_imp, cols_cat_imp, metodo='mediana')
print(f'Faltantes post-imputacion: {df.isna().sum().sum()}')
```

La funcion `imputar` rellena nulos numericos con mediana y categoricos con la moda. El patron de faltantes es MAR, por lo que la imputacion por mediana/moda es apropiada.

<font face="Times New Roman">

`test_normalidad_pre_post` evalua como afectan tres transformaciones (`log`, Box-Cox, Yeo-Johnson) la normalidad de la variable respuesta.
</font>

```python
def test_normalidad_pre_post(s):
    s = s.dropna()
    resultados = {}
    _, p0 = stats.normaltest(s) if len(s) > 5000 else stats.shapiro(s[:5000])
    resultados['original'] = p0
    s_pos = s[s > 0]
    if len(s_pos) > 0:
        s_log = np.log(s_pos)
        _, p_log = stats.normaltest(s_log) if len(s_log) > 5000 else stats.shapiro(s_log[:5000])
        resultados['log'] = p_log
    if (s > 0).all():
        try:
            s_bc, _ = stats.boxcox(s)
            _, p_bc = stats.normaltest(s_bc) if len(s_bc) > 5000 else stats.shapiro(s_bc[:5000])
            resultados['box-cox'] = p_bc
        except: resultados['box-cox'] = 0.0
    try:
        pt = PowerTransformer(method='yeo-johnson')
        s_yj = pt.fit_transform(s.values.reshape(-1,1)).ravel()
        _, p_yj = stats.normaltest(s_yj) if len(s_yj) > 5000 else stats.shapiro(s_yj[:5000])
        resultados['yeo-johnson'] = p_yj
    except: resultados['yeo-johnson'] = 0.0
    return resultados

res_transf = test_normalidad_pre_post(y)
print('p-values normalidad para monto:')
for k, v in res_transf.items():
    print(f'  {k:14s}  p = {v:.3e}')
best_transf = max(res_transf, key=res_transf.get)
print(f'\nMejor transformacion: {best_transf} (p = {res_transf[best_transf]:.3e})')
```

<font face="Times New Roman">

La funcion `construir_features` agrega las features derivadas con justificacion actuarial:

1. **`densidad_estado`:** Frecuencia relativa de creditos por estado — proxy de actividad inmobiliaria regional.
2. **`log_acciones`:** Logaritmo del numero de acciones crediticias — reduce la influencia de valores extremos.
3. **`edad_midpoint`:** Punto medio del rango de edad del acreditado — la edad influye en el perfil crediticio.
</font>

```python
def construir_features(df):
    df = df.copy()
    if 'estado' in df.columns:
        cnt = df['estado'].value_counts()
        df['densidad_estado'] = df['estado'].map(cnt).astype(float) / len(df)
        print(f'  densidad_estado: rango [{df["densidad_estado"].min():.4f}, {df["densidad_estado"].max():.4f}]')
    if 'acciones' in df.columns:
        df['log_acciones'] = np.log1p(df['acciones'].fillna(0).astype(float))
        print(f'  log_acciones: rango [{df["log_acciones"].min():.2f}, {df["log_acciones"].max():.2f}]')
    if 'edad_midpoint' in df.columns:
        df['edad_midpoint'] = df['edad_midpoint'].fillna(df['edad_midpoint'].median())
        print(f'  edad_midpoint: rango [{df["edad_midpoint"].min():.0f}, {df["edad_midpoint"].max():.0f}]')
    return df

df = construir_features(df)
```

<font face="Times New Roman">

Para regresion, se utiliza **target encoding con suavizado bayesiano** en lugar de WoE (que es para clasificacion binaria):

$$\hat{y}_c = \frac{n_c \cdot \bar{y}_c + m \cdot \bar{y}_{\text{global}}}{n_c + m}$$

**Nota R:** en R, los factores se manejan nativamente con `factor()` y los modelos como `glm` y `rpart` los tratan automaticamente.
</font>

```python
def target_encode(series, y_vals, smoothing=10):
    global_mean = y_vals.mean()
    df_ = pd.DataFrame({'x': series.astype(str), 'y': y_vals})
    grp = df_.groupby('x')['y'].agg(['mean', 'count'])
    grp['smooth'] = (grp['count'] * grp['mean'] + smoothing * global_mean) / (grp['count'] + smoothing)
    mapping = grp['smooth'].to_dict()
    encoded = series.astype(str).map(mapping).fillna(global_mean)
    return encoded, mapping

te_maps = {}
for c in cols_cat:
    if c in df.columns:
        df[c + '_te'], te_maps[c] = target_encode(df[c], y, smoothing=10)
        print(f'  {c}_te: min={df[c+"_te"].min():,.0f}, max={df[c+"_te"].max():,.0f}')
print(f'\nTarget encoding aplicado a {len(te_maps)} variables categoricas.')
```

<font face="Times New Roman">

**Estandarizacion selectiva:**
- **SVM, KNN, Redes Neuronales:** requieren estandarizacion porque sus funciones de distancia/gradientes son sensibles a la escala.
- **Arboles de decision:** los splits son **invariantes a transformaciones monotonas** (incluyendo estandarizacion).
- **MLG (TweedieRegressor):** la estandarizacion ayuda a la convergencia numerica.
</font>

```python
FEATURES_BASE = []
if 'edad_midpoint' in df.columns: FEATURES_BASE.append('edad_midpoint')
if 'anio' in df.columns: FEATURES_BASE.append('anio')

FEATURES_DERIVED = []
for c in ['densidad_estado', 'log_acciones']:
    if c in df.columns: FEATURES_DERIVED.append(c)

FEATURES_TE = [c + '_te' for c in cols_cat if c + '_te' in df.columns]
FEATURES = FEATURES_BASE + FEATURES_DERIVED + FEATURES_TE
FEATURES = [f for f in FEATURES if f in df.columns]

for c in FEATURES:
    if df[c].isna().any(): df[c] = df[c].fillna(df[c].median())

X = df[FEATURES].copy()
print(f'Features finales ({len(FEATURES)}): {FEATURES}')
print(f'X: {X.shape}  |  y: {y.shape}  |  mean(monto): {y.mean():,.0f}')
```

## Parte 3 — Particion y Validacion Cruzada ``

La validacion cruzada con $k$ folds estima el error de generalizacion con menor sesgo que un solo hold-out. Sea $\hat{R}_{hold}$ la estimacion del riesgo con un solo split. Su sesgo es:

$$\text{Bias}(\hat{R}_{hold}) = R(f_{n_{train}}) - R(f_n) > 0$$

Con CV de $k$ folds, promediamos $k$ estimaciones cada una entrenada con $n(1-1/k)$ datos:

$$\hat{R}_{CV}^{(k)} = \frac{1}{k} \sum_{j=1}^{k} \hat{R}_j$$

A medida que $k \to n$ (leave-one-out), el sesgo tiende a cero. El compromiso estandar es $k \in \{5, 10\}$.
</font>

```python
resultados_split = []
for prop in [0.70, 0.75, 0.80]:
    Xtr, Xte, ytr, yte = train_test_split(X, y, train_size=prop, random_state=RANDOM_STATE)
    sc = StandardScaler().fit(Xtr)
    m = Ridge(alpha=1.0)
    m.fit(sc.transform(Xtr), np.log(ytr))
    pred = np.exp(m.predict(sc.transform(Xte)))
    rmse = np.sqrt(mean_squared_error(yte, pred))
    r2 = r2_score(yte, pred)
    resultados_split.append({'prop_train': prop, 'n_train': len(Xtr),
                             'n_test': len(Xte), 'RMSE_test': round(rmse, 0), 'R2_test': round(r2, 4)})
display(pd.DataFrame(resultados_split))
```

<font face="Times New Roman">

Se compararon tres proporciones de train/test (70/30, 75/25, 80/20). Las diferencias son minimas. Mantenemos **75/25** como compromiso estandar.
</font>

<font face="Times New Roman">

No se requiere balanceo de clases (problema de regresion). En su lugar, comparamos el efecto de la transformacion de la respuesta.

```python
Xtr, Xte, ytr, yte = train_test_split(X, y, train_size=0.75, random_state=RANDOM_STATE)
sc = StandardScaler().fit(Xtr)

print('Comparacion de transformaciones de la respuesta (OLS):')
print(f'{"Transf":14s} {"RMSE":>12s} {"MAE":>12s} {"R2":>8s}')
print('-' * 50)

for name, t_fn, inv_fn in [
    ('none',  lambda yy: yy,        lambda p: p),
    ('log',   lambda yy: np.log(yy), lambda p: np.exp(p)),
]:
    lr = LinearRegression()
    lr.fit(sc.transform(Xtr), t_fn(ytr))
    pred = inv_fn(lr.predict(sc.transform(Xte)))
    pred = np.clip(pred, 1, None)
    rmse = np.sqrt(mean_squared_error(yte, pred))
    mae = mean_absolute_error(yte, pred)
    r2 = r2_score(yte, pred)
    print(f'{name:14s} {rmse:>12,.0f} {mae:>12,.0f} {r2:>8.4f}')

try:
    pt = PowerTransformer(method='yeo-johnson')
    ytr_yj = pt.fit_transform(ytr.values.reshape(-1,1)).ravel()
    lr = LinearRegression()
    lr.fit(sc.transform(Xtr), ytr_yj)
    pred_yj = pt.inverse_transform(lr.predict(sc.transform(Xte)).reshape(-1,1)).ravel()
    pred_yj = np.clip(pred_yj, 1, None)
    rmse = np.sqrt(mean_squared_error(yte, pred_yj))
    mae = mean_absolute_error(yte, pred_yj)
    r2 = r2_score(yte, pred_yj)
    print(f'{"yeo-johnson":14s} {rmse:>12,.0f} {mae:>12,.0f} {r2:>8.4f}')
except Exception as e:
    print(f'yeo-johnson fallo: {e}')
```

## Parte 4 — Modelos `[50 pts = 10 x 5]`

Construimos una funcion que evalua el producto cartesiano de configuraciones y selecciona automaticamente el ganador por $\lambda$-score:

$$\lambda_{\text{score}} = (1 - \lambda) \cdot (-\text{RMSE}_{CV}) - \lambda \cdot |(-\text{RMSE}_{train}) - (-\text{RMSE}_{CV})|$$
</font>

```python
from itertools import product
from time import perf_counter

def pipeline_experimentos_reg(estimator_factory, param_grid, X, y,
                               props=(0.75,), transf_y_options=('none', 'log'),
                               estandarizar=True, k_cv=5, lam=0.5,
                               max_configs=None, verbose=False):
    keys = list(param_grid.keys())
    grids = list(product(*[param_grid[k] for k in keys]))
    combos = list(product(grids, props, transf_y_options))
    if max_configs: combos = combos[:max_configs]
    rows = []
    kf = KFold(n_splits=k_cv, shuffle=True, random_state=RANDOM_STATE)

    for idx_combo, (gridvals, prop, transf_y) in enumerate(combos):
        params = dict(zip(keys, gridvals))
        Xtr, Xte, ytr, yte = train_test_split(X, y, train_size=prop, random_state=RANDOM_STATE)

        if transf_y == 'log':
            ytr_t = np.log(ytr.values) if hasattr(ytr, 'values') else np.log(ytr)
            inv_fn = np.exp
        elif transf_y == 'yeo-johnson':
            _pt = PowerTransformer(method='yeo-johnson')
            _ytr_arr = ytr.values.reshape(-1,1) if hasattr(ytr, 'values') else np.array(ytr).reshape(-1,1)
            ytr_t = _pt.fit_transform(_ytr_arr).ravel()
            inv_fn = lambda p, _pt=_pt: _pt.inverse_transform(np.array(p).reshape(-1,1)).ravel()
        else:
            ytr_t = ytr.values if hasattr(ytr, 'values') else np.array(ytr)
            inv_fn = lambda p: np.array(p)

        if estandarizar:
            sc = StandardScaler().fit(Xtr)
            Xtr_ = sc.transform(Xtr); Xte_ = sc.transform(Xte)
        else:
            Xtr_ = Xtr.values if hasattr(Xtr, 'values') else np.array(Xtr)
            Xte_ = Xte.values if hasattr(Xte, 'values') else np.array(Xte)

        t0 = perf_counter()
        try:
            est = estimator_factory(**params)
            valid_params = {k: v for k, v in params.items() if k in est.get_params()}
            cv_rmses = []
            ytr_arr = ytr.values if hasattr(ytr, 'values') else np.array(ytr)

            for tr_idx, va_idx in kf.split(Xtr_):
                Xa, Xb = Xtr_[tr_idx], Xtr_[va_idx]
                ya = ytr_t[tr_idx]
                yb_orig = ytr_arr[va_idx]
                reg = estimator_factory(**valid_params)
                reg.fit(Xa, ya)
                pred_t = reg.predict(Xb)
                pred_orig = np.clip(inv_fn(pred_t), 1, None)
                cv_rmses.append(np.sqrt(mean_squared_error(yb_orig, pred_orig)))

            cv_mean = np.mean(cv_rmses); cv_std = np.std(cv_rmses)
            reg_final = estimator_factory(**valid_params)
            reg_final.fit(Xtr_, ytr_t)
            pred_tr = np.clip(inv_fn(reg_final.predict(Xtr_)), 1, None)
            pred_te = np.clip(inv_fn(reg_final.predict(Xte_)), 1, None)
            yte_arr = yte.values if hasattr(yte, 'values') else np.array(yte)
            rmse_tr = np.sqrt(mean_squared_error(ytr_arr, pred_tr))
            rmse_te = np.sqrt(mean_squared_error(yte_arr, pred_te))
            mae_te = mean_absolute_error(yte_arr, pred_te)
            r2_tr = r2_score(ytr_arr, pred_tr)
            r2_te = r2_score(yte_arr, pred_te)
            t_elapsed = perf_counter() - t0
            lam_score = (1-lam)*(-cv_mean) - lam*abs((-rmse_tr) - (-cv_mean))
            rows.append({**params, 'prop_train': prop, 'transf_y': transf_y,
                         'RMSE_train': rmse_tr, 'R2_train': r2_tr,
                         'RMSE_CV_mean': cv_mean, 'RMSE_CV_std': cv_std,
                         'RMSE_test': rmse_te, 'MAE_test': mae_te, 'R2_test': r2_te,
                         't_s': t_elapsed, 'lambda_score': lam_score})
        except Exception as e:
            if verbose: print(f'FAIL config {idx_combo}: {params} | {e}')

    res = pd.DataFrame(rows).sort_values('lambda_score', ascending=False).reset_index(drop=True)
    res['Ganador'] = ''
    if len(res) > 0: res.loc[0, 'Ganador'] = 'X'
    return res
```

<font face="Times New Roman">

La funcion `pipeline_experimentos_reg` evalua todas las combinaciones de hiperparametros, proporciones de split y transformaciones de la respuesta. Selecciona el ganador por $\lambda_{0.5}$-score.
</font>

<font face="Times New Roman">

### 4.1 MLG Gamma / Tweedie

**La familia Gamma pertenece a la familia exponencial.** La densidad Gamma se escribe en forma canonica:

$$f(y; \theta, \phi) = \exp\left\{\frac{y\theta - b(\theta)}{\phi} + c(y, \phi)\right\}$$

donde $\theta = -1/\mu$ es el parametro natural, $b(\theta) = -\log(-\theta)$, $\phi = 1/\alpha$ es el parametro de dispersion, $\text{Var}(Y) = \phi \mu^2$.

El **enlace logaritmico** $g(\mu) = \log(\mu)$ se prefiere porque:
1. Garantiza $\hat{\mu} > 0$ para cualquier predictor lineal.
2. Coeficientes con interpretacion multiplicativa: $e^{\beta_j}$ = factor multiplicativo sobre el monto.

**Tweedie:** $\text{Var}(Y) = \phi \mu^p$. Con $p=2$ se obtiene la Gamma, con $p=1$ la Poisson.
</font>

<font face="Times New Roman">

### 4.2 Arboles de Decision (Regresion)

Para regresion, el criterio de split busca minimizar la impureza dentro de cada nodo:

**MSE:** $\text{Impureza}(t) = \frac{1}{|t|}\sum_{i \in t}(y_i - \bar{y}_t)^2$

**MAE:** $\text{Impureza}(t) = \frac{1}{|t|}\sum_{i \in t}|y_i - \text{mediana}_t|$

MAE es mas robusto a outliers — importante para `monto` con cola derecha pesada.

**Equivalencia Gini-Entropia (caso binario):** Para clasificacion binaria con $p = P(Y=1|t)$: Gini $G(p) = 2p(1-p)$, Entropia $H(p) = -p\log p - (1-p)\log(1-p)$. Ambas son concavas, simetricas en $p=0.5$ y con el mismo punto de maxima impureza. Minimizar una equivale a minimizar la otra.
</font>

### 4.3 SVR (Support Vector Regression)

El SVR con perdida $\varepsilon$-insensible resuelve:

$$\min_{w,b,\xi,\xi^*} \tfrac{1}{2}\|w\|^2 + C\sum_{i=1}^n (\xi_i + \xi_i^*)$$

sujeto a $|y_i - (w^\top x_i + b)| \le \varepsilon + \xi_i$. Los errores dentro del tubo de ancho $2\varepsilon$ no se penalizan.

**Problema dual** con multiplicadores de Lagrange $\alpha_i, \alpha_i^*$: el **kernel trick** permite operar en un espacio de dimension superior sin computar explicitamente la transformacion $\phi(x)$: basta con evaluar $K(x_i, x_j) = \phi(x_i)^\top\phi(x_j)$. Para el kernel RBF, $K(x,x') = \exp(-\gamma\|x-x'\|^2)$, el espacio inducido es de dimension infinita.
</font>

<font face="Times New Roman">

### 4.4 K-Vecinos mas Cercanos (Regresion)

**Cota de Cover-Hart (adaptada a regresion).** Para $n \to \infty$ y $k=1$, el estimador 1-NN satisface:

$$E[(\hat{f}_{1NN}(x) - f(x))^2] \le 2\sigma^2(x)$$

donde $\sigma^2(x) = \text{Var}(Y|X=x)$ es la varianza del ruido. En clasificacion, esto es $R_{1NN} \le 2R^*$.

Para regresion con $k > 1$:
$$\hat{f}_k(x) = \frac{1}{k}\sum_{i \in N_k(x)} y_i \quad \text{(pesos uniformes)}$$
$$\hat{f}_k(x) = \frac{\sum_{i \in N_k(x)} w_i y_i}{\sum w_i}, \quad w_i = 1/d(x, x_i) \quad \text{(pesos por distancia)}$$
</font>

<font face="Times New Roman">

La clase `KNNRegressorFromScratch` implementa desde cero el algoritmo KNN para regresion. Soporta distancias euclideana, Manhattan y Minkowski. Se valida contra sklearn. **(+5 pts bonus)**
</font>

```python
class KNNRegressorFromScratch:
    def __init__(self, n_neighbors=5, metric='euclidean', p=2, weights='uniform'):
        self.n_neighbors = n_neighbors; self.metric = metric; self.p = p; self.weights = weights

    def fit(self, X, y):
        self.X_ = np.asarray(X, dtype=np.float64)
        self.y_ = np.asarray(y, dtype=np.float64)
        return self

    def _distancias(self, Xq):
        if self.metric == 'euclidean':
            return np.sqrt(((Xq[:,None,:] - self.X_[None,:,:]) ** 2).sum(-1))
        elif self.metric == 'manhattan':
            return np.abs(Xq[:,None,:] - self.X_[None,:,:]).sum(-1)
        elif self.metric == 'minkowski':
            return (np.abs(Xq[:,None,:] - self.X_[None,:,:]) ** self.p).sum(-1) ** (1/self.p)
        raise ValueError(f'Metrica no soportada: {self.metric}')

    def predict(self, Xq):
        Xq = np.asarray(Xq, dtype=np.float64)
        out = np.zeros(len(Xq)); k = self.n_neighbors; bloque = 500
        for i in range(0, len(Xq), bloque):
            batch = Xq[i:i+bloque]; D = self._distancias(batch)
            idx_k = np.argpartition(D, k, axis=1)[:, :k]
            for j in range(len(batch)):
                vals = self.y_[idx_k[j]]
                if self.weights == 'uniform': out[i+j] = vals.mean()
                else:
                    dists = D[j, idx_k[j]]; w = 1.0/(dists+1e-9)
                    out[i+j] = np.average(vals, weights=w)
        return out

    def get_params(self, deep=True):
        return {'n_neighbors': self.n_neighbors, 'metric': self.metric, 'p': self.p, 'weights': self.weights}

    def set_params(self, **params):
        for k, v in params.items(): setattr(self, k, v)
        return self

# Validacion contra sklearn
Xv = StandardScaler().fit_transform(X.iloc[:3000])
yv = y.iloc[:3000].values; n_val = 300

knn_scratch = KNNRegressorFromScratch(n_neighbors=7, metric='euclidean', weights='uniform')
knn_scratch.fit(Xv[:-n_val], yv[:-n_val])
pred_s = knn_scratch.predict(Xv[-n_val:])

knn_sk = KNeighborsRegressor(n_neighbors=7, metric='euclidean', weights='uniform')
knn_sk.fit(Xv[:-n_val], yv[:-n_val])
pred_sk = knn_sk.predict(Xv[-n_val:])

max_diff = np.max(np.abs(pred_s - pred_sk))
print(f'KNNFromScratch  RMSE: {np.sqrt(mean_squared_error(yv[-n_val:], pred_s)):>12,.0f}')
print(f'sklearn KNN     RMSE: {np.sqrt(mean_squared_error(yv[-n_val:], pred_sk)):>12,.0f}')
print(f'Max diff predicciones: {max_diff:.6f}')
print(f'Concordancia: {"100%" if max_diff < 0.01 else f"diff = {max_diff:.4f}"}')
```

### 4.5 Red Neuronal — MLP Shallow (Regresion)

**Teorema de aproximacion universal (Cybenko 1989, Hornik 1991).** Sea $\sigma$ una activacion continua no constante y acotada. Para cualquier $f: [0,1]^n \to \mathbb{R}$ continua y $\varepsilon > 0$, existe un MLP con una capa oculta que satisface $\sup_x |F(x) - f(x)| < \varepsilon$.

Para regresion, la capa de salida tiene un solo neurona con activacion identidad: $\hat{y} = w_o^\top h + b_o$.

**Backpropagation (una capa oculta).** Sea $\mathcal{L} = \frac{1}{2}(\hat{y} - y)^2$. Con $h = \sigma(W^{(1)}x + b^{(1)})$ y $\hat{y} = W^{(2)}h + b^{(2)}$:

$$\frac{\partial \mathcal{L}}{\partial W^{(1)}} = (\hat{y} - y) \cdot W^{(2)\top} \circ \sigma'(W^{(1)}x + b^{(1)}) \cdot x^\top$$
</font>

```python
p = X.shape[1]
grid_mlp = {
    'hidden_layer_sizes': [(p,), (int(1.5*p),), (2*p,), (p, p//2)],
    'activation': ['relu', 'tanh'],
    'alpha': [1e-4, 1e-3, 1e-2],
    'learning_rate_init': [1e-3],
}
def fab_mlp(**kw):
    base = dict(max_iter=500, early_stopping=True, validation_fraction=0.1, random_state=RANDOM_STATE)
    base.update(kw)
    return MLPRegressor(**base)

np.random.seed(RANDOM_STATE)
idx_mlp = np.random.choice(len(X), size=min(40_000, len(X)), replace=False)
X_mlp = X.iloc[idx_mlp].reset_index(drop=True)
y_mlp = y.iloc[idx_mlp].reset_index(drop=True)

res_mlp = pipeline_experimentos_reg(fab_mlp, grid_mlp, X_mlp, y_mlp,
                                     props=(0.75, 0.80), transf_y_options=('log',),
                                     estandarizar=True, k_cv=3, lam=0.5, max_configs=30)
print(f'Configuraciones evaluadas: {len(res_mlp)}')
display(res_mlp.head(15).round(4))
```

```python
C_UNDER = 0.10; C_OVER = 0.05

def costo_asimetrico(y_true, y_pred, c_under=C_UNDER, c_over=C_OVER):
    residuals = y_true - y_pred
    cost_under = np.maximum(residuals, 0) * c_under
    cost_over = np.maximum(-residuals, 0) * c_over
    return (cost_under + cost_over).mean()

print(f'Costos: C_subestimacion = {C_UNDER:.0%}, C_sobreestimacion = {C_OVER:.0%}\n')
costos = {n: costo_asimetrico(yte_arr, pred) for n, pred in preds_te.items()}
costo_df = pd.DataFrame({'Metodo': costos.keys(),
    'Costo_asimetrico_medio_$': [round(v,0) for v in costos.values()]}).sort_values('Costo_asimetrico_medio_$')
display(costo_df)
```

```python
def diebold_mariano(y_true, pred1, pred2, h=1):
    e1 = (y_true - pred1)**2; e2 = (y_true - pred2)**2
    d = e1 - e2; d_bar = d.mean(); T = len(d)
    gamma_0 = np.var(d, ddof=1); gamma_sum = 0.0
    for k_lag in range(1, h):
        gamma_k = np.cov(d[:-k_lag], d[k_lag:])[0,1]
        gamma_sum += 2*gamma_k
    var_d = (gamma_0 + gamma_sum) / T
    if var_d <= 0: return d_bar, np.nan
    dm_stat = d_bar / np.sqrt(var_d)
    return dm_stat, 2*(1 - stats.norm.cdf(abs(dm_stat)))

print('Test de Diebold-Mariano (H0: igual capacidad predictiva)\n')
pairs = []
names = list(preds_te.keys())
for i, a in enumerate(names):
    for b in names[i+1:]:
        dm, pv = diebold_mariano(yte_arr, preds_te[a], preds_te[b])
        pairs.append({'A': a, 'B': b,
            'RMSE_A': round(np.sqrt(mean_squared_error(yte_arr, preds_te[a])),0),
            'RMSE_B': round(np.sqrt(mean_squared_error(yte_arr, preds_te[b])),0),
            'DM_stat': round(dm,3),
            'p_value': round(pv,4) if not np.isnan(pv) else None,
            'Significativo': 'Si' if (not np.isnan(pv) and pv < 0.05) else 'No'})
display(pd.DataFrame(pairs))
```

### Conclusiones

**Modelo recomendado para produccion: MLG Gamma con enlace log.**

Justificacion:
1. **Interpretabilidad directa:** los coeficientes $e^{\beta_j}$ se leen como factores multiplicativos sobre el monto.
2. **Fundamento estadistico solido:** la familia Gamma con enlace log modela correctamente la distribucion positiva y sesgada de los montos.
3. **Estabilidad:** gap train/CV minimo.
4. **Costo computacional irrisorio.**
5. **Regulacion:** los modelos actuariales en Mexico requieren transparencia y auditabilidad.

**Por que no los otros:**
- **MLP:** puede dar mejor R2 pero es una caja negra, entrena mucho mas lento.
- **SVR:** limitado por la complejidad $O(n^2)$ y la necesidad de subsamplear.
- **KNN:** sensible a la dimensionalidad; costo de prediccion alto.
- **Arbol:** alta varianza; predicciones constantes por region.

**Limitaciones:**
- Techo de R2 modesto: faltan variables como historial crediticio e ingreso del acreditado.
- Target encoding con leakage potencial.
- Validacion temporal seria mas realista.
