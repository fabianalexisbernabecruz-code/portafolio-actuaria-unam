Los dueños de ésta necesitan información cuantitativa para decidir si es razonable adquirir más tiendas durante el próximo año.


<font face="Times New Roman">

```r
# Cargar datos
url_datos_greens <- "https://raw.githubusercontent.com/fabianalexisbernabecruz-code/Modelos-Lineales/1462cbeecb0d6eb5d7dc390f30b7c977f7030f5b/Tarea_1/greens.csv"
datos_greens <- read_csv(url_datos_greens)

# Renombrar columnas
colnames(datos_greens) <- c("ventas", "pies_cuadrados", "inventario",
                            "publicidad", "tamano_distrito", "competencia")
head(datos_greens)
```

## (a) Ajuste un modelo de regresión lineal relacionando las ventas anuales netas con los pies cuadrados, inventario, gastos en publicidad, tamaño del distrito y número de negocios de la competencia. Verifique supuestos del modelo, en caso de que alguno no se cumpla indique cuál y corríjalo.


```r
# Ajustar modelo completo
modelo_completo <- lm(ventas ~ pies_cuadrados + inventario + publicidad +
                      tamano_distrito + competencia, data = datos_greens)
summary(modelo_completo)

# Verificación de supuestos

# Linealidad Gráfico de residuos vs valores ajustados
par(mfrow = c(2, 2))
plot(modelo_completo)

# Normalidad de residuos
shapiro.test(residuals(modelo_completo))
qqnorm(residuals(modelo_completo))
qqline(residuals(modelo_completo))

# Homocedasticidad
bptest(modelo_completo)

# Independencia
dwtest(modelo_completo)
```

<font face="Times New Roman">



Verifiquemos si los supuestos del modelo se cumplen:

1. Normalidad de los residuos

*   Shapiro-Wilk p-valor = 0.03651 (< 0.05)
*   Esto indica que los residuos NO siguen una distribución normal
*   No se cumple el supuesto de normalidad de los residuos

2. Homocedasticidad (varianza constante)

*   Breusch-Pagan p-valor = 0.9394 (> 0.05)
*   No hay evidencia de heterocedasticidad
*   La varianza de los residuos es constante

3. Independencia de los residuos

*   Durbin-Watson p-valor = 0.2202 (> 0.05)
*   No hay autocorrelación significativa
*   Los residuos son independientes


</font>

<font face="Times New Roman">

Corrijamos el supuesto de normalidad:

</font>

```r
#Transformación Box-Cox
boxcox_result <- boxcox(modelo_completo, lambda = seq(-2, 2, 0.1))
# El lambda óptimo es el que maximiza la log-verosimilitud
lambda_optimo <- boxcox_result$x[which.max(boxcox_result$y)]
cat("Lambda óptimo para transformación Box-Cox:", lambda_optimo, "\n")

# Aplicar transformación Box-Cox
if(abs(lambda_optimo) < 0.01) {
  # Si lambda es cercano a 0, usar log
  datos_greens$ventas_trans <- log(datos_greens$ventas)
} else {
  # Transformación general Box-Cox
  datos_greens$ventas_trans <- (datos_greens$ventas^lambda_optimo - 1)/lambda_optimo
}

# Ajustar modelo con variable transformada
modelo_boxcox <- lm(ventas_trans ~ pies_cuadrados + inventario + publicidad +
                    tamano_distrito + competencia, data = datos_greens)

# Verificar normalidad
shapiro.test(residuals(modelo_boxcox))
```

<font face="Times New Roman">

Ahora tenemos:

* Shapiro-Wilk p-valor = 0.07167 (> 0.05)
* Ya se cumple el supuesto de normalidad



</font>

<font face="Times New Roman">

---
</font>

<font face="Times New Roman">

## (b) Interprete el ajuste del modelo del inciso anterior.


```r
summary(modelo_boxcox)
```

<font face="Times New Roman">

Modelo ajustado (con transformación Box-Cox, λ = 0.424):

**Ventas_trans = −14.33 + 22.68 X2 + 0.30 X3 + 15.22 X4 + 18.42 X5 −9.46 X6**

## (h) Se mencionó que en R, la ANOVA incluye pruebas condicionales. Ajuste la regresión con solamente $X2$ y $X3$ como variables significativas. Verifique que en el segundo renglón de la ANOVA se está haciendo la prueba para ver si $X3$ es significativa, mientras que en el primer renglón se tiene la prueba para ver si $X2$ es significativa dado que $X3$ está en el modelo. Para esto último necesita obtener $SE_{H_0:\beta_2=0}(X2|X3)= SE_{H_0:\beta_2=\beta_3=0}(X2, X3) - SE_{H_0:\beta_3=0}(X2, X3)$, la primera corresponde a la suma explicada para ver si conjuntamente las dos variables explicativas son significativas, mientras que el segundo es la suma explicada para ver si $X3$ es significativa.


```r
# Ajustar modelo reducido con solo pies_cuadrados (X2) e inventario (X3)
modelo_reducido <- lm(ventas_trans ~ pies_cuadrados + inventario, data = datos_greens)

# Ver resumen del modelo
summary(modelo_reducido)

# Tabla ANOVA del modelo reducido
anova_reducido <- anova(modelo_reducido)
print(anova_reducido)

cat("\n=== SUMA DE CUADRADOS SECUENCIALES ===\n\n")

# Sumas de Cuadrados(sc) del modelo completo con X2 y X3
sc_x2_x3 <- sum(anova_reducido$`Sum Sq`[1:2])
cat("SC(X2, X3) conjunta:", sc_x2_x3, "\n")

# Sumas de Cuadrados de X3 solo (modelo con solo inventario)
modelo_x3 <- lm(ventas_trans ~ inventario, data = datos_greens)
sc_x3 <- anova(modelo_x3)$`Sum Sq`[1]
cat("SC(X3) sola:", sc_x3, "\n")

# Sumas de Cuadrados de X2 dado que X3 ya está en el modelo
# SEH0:β2=0(X2|X3) = SEH0:β2=β3=0(X2, X3) − SEH0:β3=0(X2, X3)
sc_x2dadox3 <- sc_x2_x3 - sc_x3
cat("\nSC(X2 | X3) = SC(X2, X3) - SC(X3) =", sc_x2_x3, "-", sc_x3, "=", sc_x2dadox3, "\n")

# Verificar que coincide con el primer renglón del ANOVA cuando X3 entra primero
# Para ver el orden inverso, ajustamos modelo con X3 primero y X2 después
modelo_orden_inverso <- lm(ventas_trans ~ inventario + pies_cuadrados, data = datos_greens)
anova_inverso <- anova(modelo_orden_inverso)
cat("\nANOVA con orden inverso (inventario primero, luego pies_cuadrados):\n")
print(anova_inverso)

# Explicación final
cat("\n=== CONCLUSIÓN ===\n")
cat("En el primer renglón del ANOVA del modelo reducido (con X2 y X3):\n")
cat("- Si el orden es [pies_cuadrados + inventario]: el primer renglón prueba H0: β2=0 (X2 sola)\n")
cat("- Si el orden es [inventario + pies_cuadrados]: el primer renglón prueba H0: β3=0 (X3 sola)\n")
cat("- El segundo renglón siempre prueba la variable adicional dado que la primera ya está en el modelo\n\n")

cat("Se cumple:\n")
cat("SC(X2 | X3) = SC(X2, X3) - SC(X3) =", sc_x2dadox3, "\n")
cat("Este valor corresponde a la suma de cuadrados de X2 después de haber incluido X3.\n")
```

<font face="Times New Roman">


Con la función anova() calculamos sumas de cuadrados secuenciales (Tipo I), donde el orden de las variables importa.

En el primer modelo (pies_cuadrados + inventario), los resultados muestran que $SC(X_2) = 1,651,090$ y la suma condicional $SC(X_3|X_2) = 269,980$.
Al invertir el orden (inventario + pies_cuadrados), obtenemos que $SC(X_3) = 1,857,762$ y la suma condicional buscada $SC(X_2|X_3) = 63,308$.
Podemos verificar la identidad de la suma de cuadrados conjunta:

$SC(X_2, X_3) = SC(X_2) + SC(X_3|X_2) = 1,651,090 + 269,980 = 1,921,070$ $SC(X_2, X_3) = SC(X_3) + SC(X_2|X_3) = 1,857,762 + 63,308 = 1,921,070$

Además, se verifica que el estadístico $F$ del segundo renglón del ANOVA invertido ($F = 10.216$) es exactamente el cuadrado del valor $t$ del coeficiente de pies_cuadrados en el resumen del modelo ($t = 3.196$, $3.196^2 = 10.214$), lo que confirma que la prueba en el ANOVA secuencial para la última variable es equivalente a la prueba de significancia individual del modelo completo (prueba condicional).
