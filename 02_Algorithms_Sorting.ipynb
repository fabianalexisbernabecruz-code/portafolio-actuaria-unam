# **Descripción de los algoritmos**

<font face="Times New Roman">

En esta sección explicaremos paso a paso el funcionamiento de cada algoritmo.

</font>

<font face="Times New Roman">

## SelectionSort

</font>

<font face="Times New Roman">

Es un algoritmo de ordenación simple y directo, a continuación detallarmeos el paso a paso.

1. El algortimo toma el primer elemento de la lista y luego busca al elemento más pequeño (o grande) en la lista no ordenada.
2. Una vez que encuentra al elemento más pequeño lo intercambia con el primer elemento de la lista.
3. Repite el paso 2, pero ahora no toma el primer elemento, puesto que ya está en su posición correcta, sino que toma al segundo de la lista y en lugar de buscar en toda la lista los hace en la que resulta después del paso (2), solo que lo busca en la que resulta de no considerar el elemento primerto pues, ya está en posición correcta y obvio siempre será el menor de la lista completa.
4. Se repiten estos pasos hasta llegar al penúltimo elemento y una vez ordenado podemos asegurar que se logra una lista ordenada.

</font>

<font face="Times New Roman">

## InsertionSort

</font>

<font face="Times New Roman">

Insertion Sort es un algoritmo que funciona para ordenar un número pequeño de elementos.

Este se ordena de izquierda a derecha, a continuación lo detallaremos paso a paso.

1. Tomamos el primer elemento de nuestra lista "x" y asumimos que está ordenado, tomamos el segundo elemento "y" y los comparamos.
2. Si el elemento "y" es menor al elemento "x", entonces movemos a "y" hacia a la izquierda, en dado caso donde sea mayor a "x" conservára su posición.
3. Una vez hecho el cambio tomamos el tercer elemento y lo comparamos con el segundo, si es menor se intercambia y se procede a volver a compar con el primero, si es menor se intercambia.
4. Este proceso se repite hasta encontrar un lugar donde el elemento seleccionado pueda encajar para cada uno de los elemento hasta el último.

* Este algoritmo es similar a cómo ordenarías cartas. A medida que revisas cada carta de la baraja y la insertas en su posición correcta comparandola con las ya ordenadas hasta que no hay niguna que no haya sido comparada.

</font>

<font face="Times New Roman">

## BubbleSort

</font>

<font face="Times New Roman">

El Bubble Sort "burbujea" los números más grandes hacia el final de la lista en cada paso. Es como hacer que los números más grandes "suban" **lentamente** a la posición correcta en la lista.

A continuación detallaremos el paso a paso.

1. Compara los primeros 2 elementos de la lista. Si el primero es más grande que el segundo se realiza el intercambio.
2. Luego se toma a la siguiente pareja de números de la derecha.
3. Se vuelve a relizar la comparación, se hace el intercambio y el número mayor queda de lado derecho hasta que se lleve el elemento mayor hasta el top de la lista (la $n-ésima$ posición).
4. Se repiten los pasos anteriores y con cada iteración la "burbuja" dismuye hasta dónde puede subir, esto hasta que quede ordenada la lista (la burbuja solo sea el primer y segundo elemento y el resto esté ordenado).

</font>

<font face="Times New Roman">

## MergeSort

</font>

<font face="Times New Roman">

MergeSort es, al igual que QuickSort, es un algoritmo recursivo que además sigue el método de "divide y vencerás" por lo tanto para su funcionamiento divide el paso a paso en 3 secciones; dividir, conquistar y combinar.

Primero veamos que si la lista a ordenar es vacía o tiene 1 solo elemento no hay qué hacer (basta con devolver la lista).
Por otra parte, si no es el caso anterior llevamos a cabo el método siguiente:

1.   *Dividir* la lista en dos sublistas tal que cada una de las sublistas tengan la misma cantidad de elementos o, a lo sumo, difieran en 1 elemento.

2.   *Conquistar* ordenando cada sublista usando el mismo algortimo (MergeSort) de forma recursiva.

3.   *Combinar* usando una función que mezcle dos sublistas ordenadas en el paso de *conquistar* esto se hace cuando se logra llegar a listas de elementos individuales y se van uniendo en listas semiordenadas y mezclando como en una baraja de cartas que tiene las de menor valor en el top de la pila y para ordenar toma el elemento menor de una pila (sublista) y lo compara con el elemento menor de la otra pila (la otra sublista) hasta que una quede vacía o ambas y producir al final el output; que es la lista ya ordenada.

</font>

<font face="Times New Roman">

## QuickSort

</font>

<font face="Times New Roman">

Este algoritmo al igual que MergeSort usa la estrategia o método divide y vencerás por ende junto con el ya mencionado algoritmo es uno de los más eficientes, es mejor que los anteriores. Describamos es paso a paso:

1. Comenzaremos dividiendo la lista por la mitad y el número que esté en medio lo llamaremos pivote (también se puede usar el primero o último de la lista como pivote).

2. Comparamos a todos los elementos contra el pivote y los elementos menores al pivote quedan de lado izquierdo y los mayores de lado derecho (en caso de haber elementos iguales al pivote se guardan en una lista junto con el pivote).

    **Vale la pena notar que los elementos iguales que el pivote o solo el pivote, según sea el caso, ya estarían en la posición correcta. Es por esto que puede ser incluso más eficiente que MergeSort, pero depende del pivote.**

3. Ahora en la nueva lista volvemos a seleccionar un nuevo pivote y repetimos el paso 1 y 2

4. Esto se realiza las veces que sean necesarias hasta que la lista quede ordenada (es similar a MergeSort).

</font>

## SelectionSort

</font>

<font face="Times New Roman">

Existen  de peores casos para Selection Sort:

- Listas ordenadas de forma ascendente; los elementos no ordenados se irán comparanado con cada número de la lista para encontrar el mínimo y luego se intercambiará con el primer elemento no ordenado (se alcanza el máximo número de comparaciones).

- Listas donde algún número se repita, esto representaría un gran problema para SelectionSort puesto que realizará un gran número de movimientos y comparaciones a pesar de no ser necesario.

* Dato Interesante:

    Selection Sort funciona encontrando repetidamente el elemento mínimo en el conjunto de datos no ordenados y moviéndolo a la posición correcta en la parte ordenada del conjunto. Por lo cual, en dado caso de tener muchos números repetidos seguiría encontrando el valor mínimo una y otra vez, realizando muchas comparaciones y movimientos aunque ambos números tengan el mismo valor.

</font>

## InsertionSort

</font>

<font face="Times New Roman">

Existen 3 peores casos para Insertion Sort.

- Cuando la lista está ordenada de mayor a menor (orden descendente), debido a que debe realizar más movimientos y comparaciones (como en el anterior), lo cual lo vuelve lento e ineficiente.

- Lista parcialmente ordenada: el hecho de que la lista este parcialente ordenada lo vuelve infeciente puesto que aún así deberá realizar una gran cantidad de comparaciones y desplazamientos innecesarios.

-  Listas muy grandes aunque este no es el peor caso no sería lo más
    recomendable para lsiats de gran longitud, existen otros algoritmos que serían mejor para este caso.

* Dato interesante:

    En algunas aplicaciones donde se requiera de una respuesta rápida,  Insertion Sort llega a ser ineficiente puesto que usa demasiados recursos y memoria y su respuesta puede variar según lo grande que sea la lista.

</font>

</font>

<font face="Times New Roman">

Los peores casos para Bubble sort.

- Elementos parcialmente ordenados de forma ascendente, ya que su complejidad será cuadrática y tendrá que realizar un mayor número de comparaciones y movimientos.

* Los elementos duplicados sin importar el orden, ocasionará un rendimiento deficiente.

* Un gran número de comparaciones e intercambios en cada iteración, ocasionará una complejidad temporal de $O(n^{2})$.

# **Clasificación de los algoritmos**

</font>

<font face="Times New Roman">

En esta sección se demuestra el orden de complejidad al que pertenece cada algoritmo usando la definición de Cota Superior Asintótica.

<font face="Times New Roman">

## SelectionSort

</font>

<font face="Times New Roman">

Como ya mencionamos en en análisis el peor caso para este algoritmo es cuando la lista está dispuesta en sentido contrario al orden que buscamos, ya que en un principio lo que queremos es encontrar el elemento más pequeño para mandarlo al inicio de la lista (suponiendo que queremos la lista de menor a mayor, cuando se quiere lo contrario es análogo) cosa que se logra tomando en primer elemento de la lista y compararlo con cada uno de los elementos de la lista, que de ser menores se hace el cambio de mínimo y como está dispuesto en orden contrario esto siempre va a suceder en la primer iteración, de hecho serán $n-1$ asignaciones (sin contar que en un principio asumimos como mínimo al rimero de la lista), en la segunda va a suceder dos veces menos que la primera y así sicesivamente $n-k$, pero lo que siempre pasa sin importar $n$ es la cantidad de comparaciones que serán $(n-1)$, esto esproduciendo así el número máximo de intercambios.

Vamos a hacer la deducción de la función de tiempo $T(n)$ del algoritmo, nos fijaremos en la cantidad de asignaciones de intercambio de minimo que hay y de comparaciones que hacer. Usaremos el **peor caso**.


Si la lista tiene $n\leq1$ elementos en este caso el agoritmo no realiza operación alguna solo se devuelve el mismo la lista, así $T(n\leq1)=0$.

Si no es el caso anterior veamos cómo se comporta $T(n)$

**$P.D$** $$T(n)\in O(n^{2})$$

Si $n=2$ tenemos en la primer iteración que recorrer la lista y hacer una sola comparacion pues solo hay dos elementos y un intercambio de mínimo y al colocar al más pequeño al inicio el otro también queda en la posición correcta, es entonces que $T(2)=1+1=2$  más 1 operación de cambio de elementos en la lista

Si $n=3$ por tratarse del **peor caso** tenemos $(3-1)$ comparaciones más 1 intercambio de lugares; el primero pasa a ser el último. Luego se pasa iterar con el segundo elemento que se harán $(3-2)$ comparaciones dado que uno de los 3 ya está ordenado, solo puede haber una comparación, por ser **el peor caso** pasa que ya está ordenado y por lo tanto no hay intercambios de lugar. Luego $T(3)=(3-1)+(3-2)+1+0$

Si $n=4$ tomamos el primero, habrá $(4-1)$ comparaciones más 1 intercambio, luego habrá $(4-2)$ compraciones más 1 intercambio, finalmente habrá $(4-3)$ comparaciones y 0 intercambios pues ya han sido ordenados.
Así $T(4)=(4-1)+(4-2)+(4-3)+1+1+0$

Podemos representar a n como la suma de dos funciones una de compraciones y otra de intercambios, tenemos lo siguiente:

$$T(n)=C(n)+I(n)$$

Donde $C(n)$ es el número de asignaciones o cambios de mínimo que ocurren, con $C(n)=n+(n-1)+(n-2)+...+1$ podemos agregar una sumando $n$ solo para simplificar un poco la suma ya que no cambia el comportamiento de lafunción. La tomamos así puesto que cada que sucede una iteración hay 1 elemento que queda ordenado,por lo tanto reduce en 1 la cantidad de cambios de mínimo que puede haber.

Y la función $I(n)$ es el número de intercambios entre el elemento de la iteración y el mínimo encontrado. Con $I(n)=n$, la podemos tomar así porque solo puede a lo más haber $n$ cambios de elementos pues como vimos para los casos particulares cada que ocurre una iteración hay solo un intercambio de elementos.
Luego
$$T(n)=\sum_{i=1}^{n}{i}+n$$

Por cálculo elemental $C(n)=\frac{n(n+1)}{2}=\frac{n^2+n}{2}=\frac{n^2}{2}+\frac{n}{2}$

Dado lo anterior $T(n)=\frac{n^2}{2}+\frac{n}{2} +n$
Tomando el término que crece más rápido y del hecho que  $\frac{n^2}{2}\leq n^2 $ podemos con ello acotar la función $T(n)$.

Asumiendo $c=1$ (o tambien tomar el un medio, realmente multiplicar por una constante no modifica el orden de complejidad ques lo que buscamos), y $g(n)=n^2$ podemos concluir que

 $$\therefore T(n)\in O(n^{2})$$

</font>

## InsertionSort

</font>

<font face="Times New Roman">

Para deducir la función de tiempo $T(n)$ que le toma al algoritmo llegar a la solución ($T(n)$ va a representar el numero de comparaciones que debe hacer el algoritmo para ordenar una lista de $n$ elementos) notemos que si la lista es vacía o tiene un solo elemento $T(n)=0$; solo devuelve la lista (pues de qué modo una lista con 1 o 0 elementos podría estár desordenada), pero no hace más.

Por otra parte, si no es el caso anterior vemos que, como mencionamos en el análisis el peor caso es cuando la lista está ordenada de forma contraria a la que lo queremos, tenemos lo siguiente:

**Para la dedución asumimos que los elementos están en orden inverso al que se quiere**.

Si la lista tiene dos elementos ($n=2$) por el proceder del algoritmo asumimos que el primer elemento ya está ordenado entonces tomamos el siguiente y lo comparamos con el anterior (1 operacion de comparación), por hipótesis debe haber un cambio de lugares, por lo tanto $T(2)=1$.

Si lista tiene 3 elementos ($n=3$) se asume el primero oredenado, se toma el segundo y se compara con el primero, luego se intercambian ($T(2)=1$), se pasa al ultimo elemento y se compara con el segundo y luego con el primero (2 comparaciones más)y se hace el intercambio, por lo anterior $T(3)=T(2)+2=3$.

Si la lista tiene 4 elementos ($n=4$) por hipótesis del **peor caso** tenemos el caso cuando $n=3$ $i.e.$ $T(3)$ dado que ordenamos primero 3 elementos y luego pasamos al cuarto donde tenemos que comparar con el tercero, luego con el segundo y finalmente con el primero (3 comparaciones) así tenemos $T(4)=T(3)+3=6$.

Dando el salto inductivo si la lista tiene $n$ elementos entonces el tiempo, que como dijimos lo tomamos del numero de operaciones, que le tomará a este algoritmo ordenar la lista es.

$$T(n)=T(n-1)+n-1$$

Que es recursiva pues a su vez $T(n-1)=T(n-2)+n-2$

Pero el proceso es finito desde que $n \in \mathbb{N}$, por lo cual podemos ver a $T(n)$  como la primer comparación del caso $n=2$ más las 2 comparaciones del caso $n=3$ así hasta la $n-1$ operación de $T(n)$ en símbolos:


$$T(n)=\sum_{i=1}^{n-1}i$$

Que, por Cálculo 1 y Gauss, es.
$$\sum_{i=1}^{n-1}i=\frac{(n-1)((n-1)+1)}{2}=\frac{(n-1)n}{2}\$$

Luego, por álgebra elemental:

$$T(n)=\frac{n(n-1)}{2}=\frac{n^2-n}{2}=\frac{n^2}{2}-\frac{n}{2}=\frac{1}{2}n^2-\frac{1}{2}n$$

Finalmente para conluir a qué orden de complejidad pertenece el algoritmo tenemos que acotar la función y basta con acotar el término cuadrático pues es el que crece más rápido y toma importancia cuando $n$ crece.


Tomemos $c=\frac{1}{2}$ y $g(n)=n^2$, entonces podemos concluir que para este algoritmo su orden de complejidad es cuadrático:

$$\therefore T(n)\in O(n^{2})$$

## BubbleSort

</font>

<font face="Times New Roman">

En cuanto al **peor caso** del algoritmo es igual que al anterior; una lista ordenada en sentido inverso al deseado debido a que es cuando el número de intercambios y comparaciones alcanza el máximo. De hecho es similar al anterior y para verlo tomemos $n=1,2,3,4$.

Caso trivial: si $n\leq1$, solo devolvemos la lista sin hacer más.

Caso contrario: $n\gt 1$

Si $n=2$ tenemos que se toma el primero y se compara con el segundo (1 operacion) para finalmente conmutarlos (1 operación más) así $T(2)=1+1=2$

Si $n=3$ tenemos el caso anterior puesto que la burbuja llegó al tope de la lista que tiene los primeros 2 elementos y necesita ahora una nueva comparación así como una conmutación (2 operaciones) pues se agregó un elemento, para finalmente en la siguiente iteración volver a repetir el paso anterior y ahora sí estar ordenada la lista, en general cuando se trata del **peor caso** un elemento para llegar al top debe realizar dos operaciones por cada $n-1$ elemento de una lista de $n$ elementos como al finalizar esta primer iteración forzosamente ya está ordenado y es inmutable lo máximo de veces que se puede hacer esto es de $n-1$ a $1$ se nos ocurre plantear esto como sigue:

$$T(n)=2\sum_{i=1}^{n-1}i+2\sum_{i=1}^{n-2}i+...+\sum_{i=1}^{2}i+1$$

Cada sumando es puede ser visto como la cantidad de operaciones ncesarias para llevar a cabo el proceso de subir la burbuja hasta el top, osea el elemento de la primera posición (el más grande de toda la lista (**peor caso**) hasta el top de la lista.

Basta con acotar la suma más grande que es la primera pues tiene un sumando mas que el segundo, 2 más que la tercera suma, etc... y claramente es más grande pues todos los sumandos de cada suma son positivos. Desarrollemos pues está primera suma Sigma.
Por álgebra y cálculo básico (al igual que las dos anteriores demostraciones):

$$2\sum_{i=1}^{n-1}i=2(\frac{n(n-1)}{2})= n^2 - n$$

Tomando el término que crece más rápido según $n$ va creciendo es: $n^2$

Asumiendo $c=1$ y $g(n)=n^2$ en la definición de cota superior asintótica concluimos.

$$\therefore T(n) \in O(n^2)$$

## MergeSort

</font>

<font face="Times New Roman">

Primero consideremos que ordenar una colección de tamaño $n$ tomará $2T(\frac{n}{2})+a\cdot n$ puesto que se parte a lista en  dos de tamaño $\frac{n}{2}$ y tomamos $b$ como el tiempo constante del caso base y $a$ de forma similar que $b$ y a esa lista de $\frac{n}{2}$ se aplica el mismo algoritmo.

Con lo que podemos definir la función de recurrencia de la siguiente forma.

$\forall n \in \mathbb{N}$
$$T(n)=\begin{cases}
b & n=0\,\acute{o}\,n=1\\
2T(\frac{n}{2})+a\cdot n & n\geq2
\end{cases} $$


**Demostración**

Usando el teorema maestro pues cumple que $a > 0$, $b > 1$ son constantes y el algoritmo divide el problema en subproblemas para resolverlos recursivamente.

Sea $T(n)$ el número de operaciones que le toma al algoritmo anterior ordenar una lista de tamaño $n$ y dada la función de recurrencia.

$P.D.$ $$T(n)=n\cdot b+a\cdot n\cdot\log_{2}n$$

Como vamos a trabajar con expresiones fraccionarias del tipo $\frac{n}{2k}$ con $k \in \mathbb{N}$
la demostración se vuelve más entendible si definimos $n=2^{k}$ y nos da la util manera de saber cuántas veces se ha dividido la lista.


$$\begin{eqnarray*}
T(n)    & = &   T(2^{k})\\
  & = & 2T(2^{k-1})+a\cdot2^{k}\\
  & = & 2(2T(2^{k-2})+a\cdot2^{k-1})+a\cdot2^{k}\\
  & = & 2^{2}T(2^{k-2})+a\cdot2^{k}+a\cdot2^{k}\\
  & = & 2^{2}(2T(2^{k-3})+a\cdot2^{k-2})+2\cdot a\cdot2^{k}\\
i-veces & \vdots & \\
  & = & 2^{i}T(2^{k-i})+i\cdot a\cdot2^{k}\\
i=k     & \vdots & \\
  & = & 2^{k}T(2^{k-k})+k\cdot a\cdot2^{k}\\
  & = & 2^{k}b+k\cdot a\cdot2^{k}\\
  & = & n\cdot b+k\cdot a\cdot n\\
  & = & n\cdot b+a\cdot n\cdot\log_{2}n\\
  \end{eqnarray*}$$

Las igualdades vienen dadas por lo siguiente:

Hay 10 igualdades.
* la primera es por conveniencia de ver las divisiones de la lista, esto es cuando definimos $n=2^k$
* la siguiente igualdad es por definición de $T(n)$,
* la tercer igualdad viene dada por la definición de $T(n)$ aplicada a $2^{k-1}$
* la cuarta por potencias de $2$ y álgebra básica
* la quinta se vuelve a aplicar la definición de $T(n)$
* Lo de $i-veces$ es porque como dijimos tomando $n$ de forma conveniente se puede ver el número de divisiones que ha hecho el algoritmo en la lista de lo que es claro el patrón queda al final del álgebra lo que sigue en la sexta igualdad
* lo de $i=k$ sucede porque haciendo esto llegamos al caso base, aquel que ya no usa la recursividad cosa que es típica de los algoritmos recursivos, se debe llegar a algo que podamos saber qué hacer y en este caso cuánto tiempo tomó.
* La igualdad 7 es reemplazo de $i=k$,
* la igualdad 8 se usa $T(1)=b$
* la novena igualdad reemplazo de $n=2^k$
* finalmente reemplazo de $k$ que se sigue de:
Si $$n=2^k \implies log_{2}{n}=log_{2}{2^k}\implies log_{2}{n}=k \cdot log_{2}{2}=k \cdot 1=k $$

$$\therefore  log_{2}{n}=k$$

Ya que $a$ y $b$ son constantes podemos concluir que, la función que 'crece' más rápido es $a\cdot n\cdot\log_{2}n$

Finalmente si tomamos $a=c$, entonces se puede concluir que

$$\therefore T(n)\in O(n\cdot\log_{2}n)$$

## SelectionSort

</font>

```python
def selectionSort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    n = len(lista)

    for i in range(n):
        # Suponemos que el primer elemento sin ordenar es el mínimo
        minimo = i

        # Buscamos el elemento más pequeño en la parte no ordenada
        for j in range(i + 1, n):
            if lista[j] < lista[minimo]:
                minimo = j

        # Intercambiamos el elemento más pequeño con el primer elemento sin ordenar
        lista[i], lista[minimo] = lista[minimo], lista[i]

    return lista

#### FIN DEL PROGRAMA
```

## InsertionSort

</font>

```python
def insertionSort(arr):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    for i in range(1, len(arr)):# Comenzamos desde el segundo elemento (índice 1)
        key = arr[i]# Guardamos el elemento actual que queremos insertar en su posición correcta
        j = i - 1# Inicializamos un índice para comparar con los elementos anteriores
                    # Comparamos el elemento actual con los elementos anteriores y desplazamos los elementos mayores hacia la derecha
        while j >= 0 and key < arr[j]:
            arr[j + 1] = arr[j]
            j -= 1
                # Colocamos el elemento actual en su posición correcta
        arr[j + 1] = key

    return arr

#### FIN DEL PROGRAMA
```

## BubbleSort

</font>

```python
# implementacion in-situ del algoritmo burbuja
def bubbleSort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    #asigno el largo de la lista a una variable
    n = len(lista)
    #bandera para identificar cambios
    intercambio = True
    #variable para saber cuantos han sido ordenados
    j = 0
    #si hay cambios entro al ciclo
    while(intercambio):
        #los j-esimos elementos ya han sido ordenados
        j = j+1
        #asumimos que no hay intercambios
        intercambio = False
        #con este for se comparan las parejas que no estan ordenadas
        for i in range(n-j):
            if lista[i] > lista[i+1] :
                # intercambio en una linea (swap)
                lista[i], lista[i+1] = lista[i+1], lista[i]
                intercambio = True
    #devuelve la lista solo para ser formales IN-SITU
    return lista

#### FIN DEL PROGRAMA
```

## MergeSort

</font>

```python
def merge_sort(lista):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    ##creamos el caso trivial, cuando no hay más qué hacer
    if len(lista) < 2:
        return lista
    ##implementamos el caso cuanto la lista tiene más de dos elemento
    else:
        medio = len(lista) // 2
        izq = merge_sort(lista[:medio])
        der = merge_sort(lista[medio:])
        return merge(izq, der)
#### FIN DEL PROGRAMA

### creamos la funcion que nos permite hacer el paso de mezclar las
## listas una vez que han sido divididas en el paso anterior
def merge(xs, ys):
    """Fusiona dos listas ordenadas en una sola lista ordenada, es una función auxiliar.

    Parameters
    ----------
    xs (list): La primera lista ordenada.
    ys (list): La segunda lista ordenada.
    xs : list
    ys : list

    Returns
    -------
     Una nueva lista que contiene todos los elementos de xs e ys, ordenados de manera ascendente : list.

    """
    ix, iy = 0, 0 # Inicializamos los índices para recorrer xs e ys.
    result = []  #Inicializamos una lista vacía para almacenar el resultado.
    while ix < len(xs) and iy < len(ys):
        if xs[ix] < ys[iy]:
            result.append(xs[ix])
            ix += 1
        else:
            result.append(ys[iy])
            iy += 1
    # Después de salir del bucle, agrega los elementos restantes (si los hay) de xs e ys.
    result += xs[ix:]
    result += ys[iy:]
    return result

#### FIN DEL PROGRAMA
```

## QuickSort

</font>

```python
def quicksort(arr):
    """
    Ordena listas

    Parameters
    ----------
    Una lista
    lista : list

    Returns
    -------
    La lista ordenada
    lista ordenada: list

    """
    # Paso base: Si la lista está vacía o tiene un solo elemento, ya está ordenada.
    if len(arr) <= 1:
        return arr

    # Elegimos un elemento pivote (generalmente el primer elemento en este caso).
    pivote = arr[0]

    # Creamos tres listas vacías para los elementos menores, iguales y mayores que el pivote.
    menor, igual, mayor = [], [], []

    # Recorremos la lista y distribuimos los elementos en las tres listas según su relación con el pivote.
    for elemento in arr:
        if elemento < pivote:
            menor.append(elemento)
        elif elemento == pivote:
            igual.append(elemento)
        else:
            mayor.append(elemento)

    # Llamamos recursivamente a quicksort en las listas menor y mayor.
    # Concatenamos las tres listas en el orden correcto para obtener la lista ordenada.
    return quicksort(menor) + igual + quicksort(mayor)

#### FIN DEL PROGRAMA
```

```python
def MedidorDeTiempos(algoritmo):

    """Es una función que toma las listas definidas antes para ordenarlas y tomar el tiempo que les toma
Retorna el tiempo que le toma a un algoritmo ordenar listas de 10, 20 y 30 elementos (las definidas previamente

Parámetros
----------
Algoritmo a se medido en tiempo
Algoritmo: Función

Returns
-------
Una terna de elementos que representan el tiempo que tomó el algoritmo para ordenar listas de 10, 20 y 30 elementos (todos toman la misma lista)
    """

    lista10_tiempo = [45, 17, 92, 61, 29, 84, 2, 73, 6, 38] #Listas del principio
    lista20_tiempo = [19, 12, 53, 62, 34, 90, 83, 46, 99, 28, 74, 41, 76, 95, 50, 59, 65, 25, 37, 16]
    lista30_tiempo = [98, 71, 20, 3, 48, 6, 79, 17, 33, 24, 93, 39, 36, 55, 86, 60, 77, 56, 1, 66, 94, 64,19, 12, 53, 62, 34, 90, 83, 46, 99, 28, 74, 41, 76, 95,98, 71, 20, 3, 48, 6, 79, 17, 33, 24, 93, 39, 36, 55, 86, 60, 77, 56, 1, 66, 94, 64,19, 12, 53, 62, 34, 50, 59, 65, 25, 37, 16, 23, 4, 89, 51, 30, 75, 13,0]

    #Creamos una copia de cada uno para ordenarla, si no lo hicieramos con un primer algoritmo la lista quedaría ordenada
    #y el resto de algoritmos ordenarían listas ya ordenadas cayendo en su peor caso y afectando la medición.
    lista10copia = lista10_tiempo.copy()
    lista20copia = lista20_tiempo.copy()
    lista30copia = lista30_tiempo.copy()

    #Usamos la función timeit
    inicio = time.time()
    algoritmo(lista10copia)
    fin = time.time()
    tiempo10 = fin - inicio

    inicio = time.time()
    algoritmo(lista20copia)
    fin = time.time()
    tiempo20 = fin - inicio

    inicio = time.time()
    algoritmo(lista30copia)
    fin = time.time()
    tiempo30 = fin - inicio
   #Devolvemos los tiempos en ternas
    return tiempo10, tiempo20, tiempo30
```
