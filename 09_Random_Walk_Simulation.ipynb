```python
import numpy as np
import matplotlib.pyplot as plt
```

<font face="Times New Roman">

Vamos a simular una caminata aleatoria simple:

$P(X_1 = n+1 | X_0 = n) = 1/2$

$P(X_1 = n-1 | X_0 = n) = 1/2$

Primero solo vamos a poder ver una parte de la cadena.
0. El estado inicial es 0 (convernción)
1. definimos que va a correr la cadena
2. en cada tiempo ver si sube o baja
3. guardar el estado en el que está la cadena (una lista)
* el espacio de estados es todo $\mathbb{Z}$
* Distribución incial $P(X_0 = 0) = 1$

</font>

```python
"""
0 baja

1 sube
"""
n = 100
estados = [0]
estados[0] = 0
np.random.binomial(n=1, p=0.5)
for i in range(1,n):
    estados.append(np.random.binomial(n=1, p=0.5))
```

```python
print(estados)
print(len(estados))
```

```python
estados2 = [0] * n
for i in range(1,n-1):
    if estados[i] == 0:
        estados2[i] = estados2[i-1]-1
    else:
        estados2[i] = estados2[i-1]+1
#salto
estados2[-1]= estados2[-2]+1
print(estados2)
```

```python
ejex = list(range(n))
plt.plot(ejex,estados2, c ="red")
plt.xlabel("Tiempo")
plt.ylabel("Estado")
plt.title("Caminata aleatoria simple")
plt.show()
```

```python
print("Partiendo del estado 0, dentro de los" , n, "pasos se llegó máximo al estado:", max(estados2), "y mínimo al estado:", min(estados2))
```

```python
#Caminata aleatoria
import numpy as np

n = 50 #Tiempo que voy a dejar correr en la simulación.
cadena = np.empty(n) #Voy a iniciar una lista vacía con el número de pasos necesario.
cadena[0] = 0 #La cadena va a iniciar en el estado 0
p = 0.5 #p es la probabilidad de que la caminata aumente en 1,

for i in range(1,n):
#En cada tiempo simulo una Bernoulli con parámetro p. Esto lo hago simulando una uniforme en (0,1) y viendo si cae en (0, p) o en (p, 1).
    uniforme = np.random.uniform()
    if uniforme < p:
        cadena[i] = cadena[i-1] + 1 #Si la Bernoulli es 1 entonces el nuevo estado es el anterior más 1.
    elif uniforme > p:
        cadena[i] = cadena[i-1] -1 #Si la Bernoulli es 0 entonces el nuevo estado es el anterior menos 1.

#Cuando acabe el for, la i-ésima entrada de la lista "cadena" tendrá el estado en en el que se encontraba la cadena al tiempo i.

print(cadena)
```
