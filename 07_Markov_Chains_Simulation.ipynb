### **1. Sea  $ \{X_n\}_{n \geq 0} $  una Cadena de Markov con espacio de estados $\{0, 1, 2, 3\}$ y matriz de probabilidades de transición:**

$$
P =
\begin{pmatrix}
1 - p & p & 0 & 0 \\
0 & 0 & 1 - p & p \\
1 - p & p & 0 & 0 \\
0 & 1 & 0 & 0
\end{pmatrix}
$$

</font>

<font face="Times New Roman">

**ii) Diseñe un algoritmo que simule la cadena.**

</font>

```python
import numpy as np

def Markov_Chain(p, steps, initial_state):
    # Definimos la matriz de transición
    P = np.array([
        [1-p, p, 0, 0],
        [0, 0, 1-p, p],
        [1-p, p, 0, 0],
        [0, 1, 0, 0]
    ])

    # Espacio de estados
    states = [0, 1, 2, 3]

    # Creamos la lista que almacenará la cadena de estados
    chain = [initial_state]

    # Simulamos la cadena
    estado_actual = initial_state
    for i in range(steps):
        # Avanzamos al siguiente estado
        estado_actual = np.random.choice(states, p=P[estado_actual])
        chain.append(estado_actual)

    return chain
```

### **2. Considere la cadena de rachas de éxitos.**

</font>

<font face="Times New Roman">

**ii) Diseñe un algoritmo que simule la cadena.**

</font>

```python
import numpy as np

def simulate_streak_chain(p, j, M):
    # Inicializar el número de visitas a cada estado
    visitas = np.zeros(3)

    # Estado inicial
    estado_actual = j

    for _ in range(M):
        # Registrar la visita
        visitas[estado_actual] += 1

        # Generar un número aleatorio
        r = np.random.rand()
        if r <= p:
            # Extiende la racha
            estado_actual = min(estado_actual + 1, 2) # Limitar a 2
        else:
            # Rompe la racha
            estado_actual = 0

    # Normalizar las visitas para obtener la frecuencia
    return visitas / M
```

<font face="Times New Roman">

**iii) Para tres distintos valores de $p$ y número de pasos $M = 100, 1000$, simule la cadena iniciando en $j$ y calcule $ \frac{N^M_j}{M} $ (donde $ N^M_j = $ número de visitas al estado $j$ hasta el tiempo $M$) para $j = 0, 1, 2$. Compare con la distribución estacionaria obtenida en (i).**

</font>

```python
import numpy as np

# Define the simulate_streak_chain function
def simulate_streak_chain(p, j, M):
    # Initialize the number of visits to each state
    visits = np.zeros(3)

    # Start at the initial state
    current_state = j

    for _ in range(M):
        # Record the visit to the current state
        visits[current_state] += 1

        # Generate a random number
        r = np.random.rand()

        if r <= p:
            # Extend the streak
            current_state = min(current_state + 1, 2)  # Limit to 2
        else:
            # Break the streak
            current_state = 0

    # Normalize the visits to get the frequency
    return visits / M

# Example parameters
p_values = [0.2, 0.5, 0.8]  # Probabilities for each simulation
M_values = [100, 1000]      # Number of steps for each simulation
initial_state = 0           # Starting state

# Simulate for each combination of p and M
resultados = {
    (p, M): simulate_streak_chain(p, initial_state, M)
    for p in p_values for M in M_values
}

# Print the results
for key, value in resultados.items():
    p, M = key
    print(f"Results for p = {p} and M = {M}:")
    print(f"State 0: {value[0]:.4f}, State 1: {value[1]:.4f}, State 2: {value[2]:.4f}\n")
```

### **3. Considere la Cadena de Ehrenfest.**

</font>

<font face="Times New Roman">

**ii) Diseñe un algoritmo para simular la cadena.**

</font>

<font face="Times New Roman">

La cadena de Ehrenfest se suele usar para modelar el intercambio aleatorio de partículas entre dos regiones. En cada tiempo, se selecciona una partícula al azar y se mueve al otro recipiente en este sentido el algoritmo debe:
1. Definir el número total de partículas $N$.
2. Inicializar el número de partículas en una urna (región), esto es, definir el estado en el que parte la cadena, digamos en el estado $j$.
3. Para cada paso de la simulación:

    *   Escoger aleatoriamente una partícula y cambiarla de recipiente.
    *   Actualizar el estado de la cadena.

4. Repetir este proceso hasta completar el número de pasos $M$.

</font>

<font face="Times New Roman">

**Implementación**

</font>

```python
import numpy as np

def CadenaEhrenfest(N, M, EstadoInicial):
    """
    Función que simula la Cadena de Ehrenfest.

    Args:
    N (int): Número total de partículas (suma de partículas de la urna A y B).
    M (int): Número de pasos de la simulación (tiempo de simulación).
    EstadoInicial (int): Estado inicial de la cadena (estado j del que parte i.e
    número de partículas en la urna (o región) A).

    Returns:
    visitas (list): Número de visitas a cada estado.
    estados (list): Secuencia de estados a lo largo del tiempo.
    """
    # Inicializamos el estado
    EstadoActual = EstadoInicial
    estados = [EstadoActual]   # Con esto almacenamos la secuencia de estados, el primero es donde partimos
    visitas = np.zeros(N + 1)  # Contamos las visitas a cada estado y lo guardamos aquí

    # Iteramos M veces (reproducimos la cadena M veces)
    for _ in range(M):
        # Escogemos aleatoriamente una partícula entre las N partículas
        particula = np.random.randint(1, N+1)  # Escoge una partícula entre 1 y N

        # Si la partícula está en la urna A (se encuentra entre las primeras 'EstadoActual' partículas)
        if particula <= EstadoActual:
            EstadoActual -= 1  # Movemos una partícula de A a B por lo que en A hay 1 partícula menos
        else:
            EstadoActual += 1  # Movemos una partícula de B a A así hay una partícula más en A

        # Guardamos el estado actual
        estados.append(EstadoActual)
        # Registramos las visitas a cada estado (es una función indicadora)
        visitas[EstadoActual] += 1

    # Devolvemos cómo se llevó a cavo el intercambio de partículas y su
    # numero total de visitas en M repeticiones del experimento
    return visitas, estados
```

**iii) Simule la cadena para $N = 3, 4$ (número total de bolas al principio del experimento) iniciando en $j$ y calcule $ \frac{N^M_j}{M} $ (donde $ N^M_j = $ número de visitas al estado $j$ hasta el tiempo $M$) para $j = 0, 1, 2$. Compare con la distribución estacionaria obtenida en (i).**

</font>

<font face="Times New Roman">

$N = 3$

</font>

```python
# Parámetros de la simulación
N = 3     # Número total de partículas
M = 1000  # Número de pasos de la simulación

# Iteramos sobre cada estado inicial j = 0, 1, 2.
for estado_inicial in range(3):
    # Simulación de la Cadena de Ehrenfest
    visitas, estados = CadenaEhrenfest(N, M, estado_inicial)

    # Resultados para el estado inicial j
    print(f"\nEstado inicial j = {estado_inicial}:")
    print(f"Número de visitas a j = {estado_inicial}: {int(visitas[estado_inicial])}")
    print(f"Frecuencia relativa de visita j = {estado_inicial}: {visitas[(estado_inicial)] / M}")
    print(f"Número de visitas a cada estado: {visitas}")
    print(f"Frecuencia relativa de visitas: {visitas / M}")
```

**Comparación con la distribución estacionaria:**

</font>

<font face="Times New Roman">

La distribución estacionaria es:

$$
\pi_i = \frac{\binom{N}{i}}{2^N}
$$

</font>

```python
from scipy.special import comb

def DistribuciónEstacionaria(N):
    """
    Calcula el vector de probabilidades pi.
    Args:
    N (int): Tamaño del vector.

    Returns:
    np.ndarray: Vector de tamaño N+1 donde cada entrada i es p(i).
    """
    pi_vector = np.zeros(N + 1)  # Inicializamos el vector de tamaño N+1
    for i in range(0, N + 1):        # j va de 0 a N
        pi_vector[i] = comb(N, i) / (2 ** N) # Calculamos pi(i)
    return pi_vector
```
